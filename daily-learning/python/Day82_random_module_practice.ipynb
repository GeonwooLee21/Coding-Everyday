## 1️⃣ `random.seed()` — 난수의 시작점 고정
### 1. `random.seed()` — Fix the random starting point

컴퓨터의 난수는 사실 **공식으로 계산된 '의사난수(pseudo random)'** 입니다.
`seed`(씨앗)를 같게 주면 **항상 똑같은 수열**이 나옵니다. → 테스트·디버깅에 필수!

**💡 예제 / Example** — 아래 셀을 실행해 결과를 확인하세요. / Run the cell below.

### 2. `random.random()` — Float in [0.0, 1.0)

# 🧩 문제 1: 의사난수(Pseudo Random) 만들기
# 🧩 Problem 1: Generating Pseudo Random Numbers

## 📖 배경 / Background

**Wonder Exchange** 테스트 팀은 매번 같은 '랜덤' 환율 데이터로 테스트하고 싶어 합니다.
"랜덤인데 매번 같다?" — 바로 **의사난수 + `seed`** 의 마법입니다!

The **Wonder Exchange** test team wants the *same* "random" rate data every test run.
"Random but always the same?" — that's the magic of **pseudo random + `seed`**!

## 🎯 미션 / Mission

함수 `make_test_rates(seed_value, count)`를 완성하세요.
Complete the function `make_test_rates(seed_value, count)`.

**규칙 / Rules**
- `seed_value`로 `random.seed()`를 고정한다 / fix the seed with `seed_value`
- `1300.0`~`1400.0` 사이 환율을 `count`개 만든다 (소수 둘째 자리 반올림) / make `count` rates in `[1300.0, 1400.0]`, rounded to 2 decimals
- 리스트로 반환한다 / return a list
- ✅ **같은 `seed_value`면 항상 같은 결과**가 나와야 한다 / same `seed_value` → identical result

> 변수·함수 이름은 모두 **snake_case** 입니다. / All names use **snake_case**.

```python
def make_test_rates(seed_value, count):
    # TODO 1: seed_value로 난수 시작점 고정 / fix the seed
    random.seed(2)
    rates = []
    for _ in range(count):
        # TODO 2: 1300.0~1400.0 환율을 둘째 자리까지 / rate in [1300, 1400], 2 decimals
        rate = round(random.uniform(1300.0, 1400.0), 2)
        rates.append(rate)
    return rates


# --- 테스트 / test ---
first  = make_test_rates(42, 5)
second = make_test_rates(42, 5)
third  = make_test_rates(2025, 5)
print("seed=42 (1회) :", first)
print("seed=42 (2회) :", second)
print("seed=2025     :", third)
print("재현성 통과? / reproducible?", first == second)        # True 여야 함
print("seed 다르면 다름? / differs?", first != third)         # True 여야 함
```

### 🔮 보너스(🥇 Hard): 직접 만드는 의사난수 — LCG
### 🔮 Bonus (🥇 Hard): Build your own — LCG

> ⚠️ 아래는 '의사난수가 어떻게 만들어지는가'를 보여주는 심화 자료입니다.
> ⚠️ Deeper material showing *how* pseudo random is actually computed.

`random` 모듈 내부도 결국 **공식 한 줄**입니다. 가장 단순한 **선형 합동 생성기(LCG)**:

The `random` module is ultimately **one formula** inside. The simplest is a **Linear Congruential Generator (LCG)**:

$$X_{n+1} = (a \cdot X_n + c) \bmod m$$

곱하기·더하기·나머지(`%`)만으로 '랜덤처럼 보이는' 수열이 나옵니다. 곱셈·나눗셈·`%`만 쓰면 됩니다!
Just multiply, add, and modulo (`%`) to get a "random-looking" sequence!

```python
def lcg_random(seed_value, count):
    # 유명한 상수 (glibc 계열) / well-known constants
    a, c, m = 1103515245, 12345, 2 ** 31
    state = seed_value
    results = []
    for _ in range(count):
        # TODO: 공식 한 줄 X = (a*X + c) % m / one-line formula
        state = (a * state + c) % m
        results.append(round(state / m, 4))   # [0,1)로 정규화 / normalize
    return results


print("LCG seed=42 :", lcg_random(42, 5))
print("LCG seed=42 :", lcg_random(42, 5))   # 똑같이 나옵니다! / identical
print("LCG seed=99 :", lcg_random(99, 5))
print("=> 컴퓨터 난수도 결국 '공식'이라 의사난수(pseudo)라 부릅니다.")
print("=> Computer randomness is just a formula, hence 'pseudo' random.")
```

## 🎯 미션 / Mission

함수 `generate_lotto_numbers()`를 완성하세요.
Complete the function `generate_lotto_numbers()`.

**규칙 / Rules**
- `1`~`45` 중 **중복 없이** 6개를 뽑는다 / pick 6 from 1–45 **without duplicates**
- **오름차순으로 정렬**해서 반환한다 / return them **sorted ascending**
- 추가 리스트를 따로 만들지 말 것 (`sample` + `sort` 활용) / use `sample` + `sort`

```python
def generate_lotto_numbers():
    # TODO 1: 1~45 중 중복 없이 6개 / 6 unique numbers from 1..45
    numbers = random.sample(range(1, 45), 6)
    # TODO 2: 오름차순 정렬 (제자리) / sort ascending in-place
    numbers.sort()
    return numbers


# --- 테스트 / test ---
for game in range(3):
    print(f"게임 {game + 1} / Game {game + 1}:", generate_lotto_numbers())
```

# 🧩 문제 3: OTP / 비밀번호 생성기
# 🧩 Problem 3: OTP / Password Generator

## 📖 배경 / Background

**Wonder Pay** 로그인 보안을 강화합니다. 로그인할 때마다 **6자리 OTP**(숫자)와,
계정 생성 시 **임시 비밀번호**(영문 + 숫자)를 만들어야 합니다.

**Wonder Pay** is hardening login security. We need a **6-digit OTP** (digits) per login,
and a **temporary password** (letters + digits) at sign-up.

> ⚠️ **실무 주의 / Real-world note**
> 실제 보안 시스템에서는 `random`이 아니라 **`secrets` 모듈**을 써야 합니다.
> `random`은 예측 가능해서 보안에 취약합니다 — 오늘은 **학습 목적**으로만 사용합니다!
> Real systems must use the **`secrets` module**, not `random`. `random` is predictable
> and unsafe for security — we use it here **for learning only**!

## 🎯 미션 / Mission

두 함수를 완성하세요. / Complete both functions.

**`generate_otp(length=6)`**
- `0`~`9` 숫자로 이루어진 `length`자리 **문자열** 반환 / return a `length`-digit string of `0`–`9`

**`generate_temp_password(length=10)`**
- 영문 대소문자 + 숫자로 이루어진 `length`자리 문자열 반환 / letters (a–z, A–Z) + digits

```python
import string   # 문자 집합을 쉽게 얻기 위해 / for handy character sets


def generate_otp(length=6):
    digits = "0123456789"
    # TODO: digits에서 length개를 복원 추출(중복 허용) / draw length chars with replacement
    otp_chars = random.choices(digits, k=length)
    # TODO: 문자 리스트를 하나의 문자열로 / join into one string
    return "".join(otp_chars)


def generate_temp_password(length=10):
    # string.ascii_letters = 'abc...XYZ', string.digits = '0123456789'
    pool = string.ascii_letters + string.digits
    chars = random.choices(pool, k=length)
    return "".join(chars)


# --- 테스트 / test ---
print("OTP        :", generate_otp(6))
print("OTP (4자리) :", generate_otp(4))
print("임시 비밀번호 / temp password:", generate_temp_password(10))
```

### 🥇 보너스(Hard): 강한 비밀번호 (각 문자 종류 최소 1개)
### 🥇 Bonus (Hard): Strong password (at least one of each type)

진짜 안전한 비밀번호는 **소문자·대문자·숫자·기호를 각각 최소 1개씩** 포함해야 합니다.
전략: 종류별로 1개씩 먼저 확보 → 나머지를 채움 → `shuffle`로 위치를 섞기!

A truly safe password should include **at least one lowercase, uppercase, digit, and symbol**.
Strategy: secure one of each first → fill the rest → `shuffle` the positions!

```python
def generate_strong_password(length=12):
    if length < 4:
        length = 4   # 최소 4자리 보장 / ensure at least 4

    lowers, uppers = string.ascii_lowercase, string.ascii_uppercase
    digits, symbols = string.digits, "!@#$%^&*"

    # TODO 1: 종류별로 최소 1개씩 확보 / one of each type
    required = [
        random.choice(lowers),
        random.choice(uppers),
        random.choice(digits),
        random.choice(symbols),
    ]
    # TODO 2: 남은 자리를 전체 풀에서 채움 / fill remaining from the full pool
    pool = lowers + uppers + digits + symbols
    rest = random.choices(pool, k=length - 4)

    password_chars = required + rest
    # TODO 3: 위치를 섞어 패턴 노출 방지 / shuffle to avoid fixed positions
    random.shuffle(password_chars)
    return "".join(password_chars)


pw = generate_strong_password(12)
print("강한 비밀번호 / strong password:", pw)
print("소문자 포함? lower :", any(c.islower() for c in pw))
print("대문자 포함? upper :", any(c.isupper() for c in pw))
print("숫자 포함?   digit :", any(c.isdigit() for c in pw))
print("기호 포함?   symbol:", any(c in "!@#$%^&*" for c in pw))
```

### 문제 1 정답 / Problem 1 Solution

```python
def make_test_rates(seed_value, count):
    random.seed(seed_value)
    rates = []
    for _ in range(count):
        rate = round(random.uniform(1300.0, 1400.0), 2)
        rates.append(rate)
    return rates

print(make_test_rates(42, 5))
print(make_test_rates(42, 5) == make_test_rates(42, 5))   # True

# 보너스 LCG
def lcg_random(seed_value, count):
    a, c, m = 1103515245, 12345, 2 ** 31
    state = seed_value
    out = []
    for _ in range(count):
        state = (a * state + c) % m
        out.append(round(state / m, 4))
    return out

print(lcg_random(42, 5))
```

```python
def generate_lotto_numbers():
    numbers = random.sample(range(1, 46), 6)
    numbers.sort()
    return numbers

def generate_lotto_games(game_count):
    games = {}
    labels = "ABCDEFGHIJ"
    for i in range(game_count):
        games[labels[i]] = generate_lotto_numbers()
    return games

print(generate_lotto_numbers())
for k, v in generate_lotto_games(5).items():
    print(k, v)
```

### 문제 3 정답 / Problem 3 Solution

```python
def generate_otp(length=6):
    digits = "0123456789"
    otp_chars = random.choices(digits, k=length)
    return "".join(otp_chars)

def generate_temp_password(length=10):
    pool = string.ascii_letters + string.digits
    return "".join(random.choices(pool, k=length))

def generate_strong_password(length=12):
    if length < 4:
        length = 4
    lowers, uppers = string.ascii_lowercase, string.ascii_uppercase
    digits, symbols = string.digits, "!@#$%^&*"
    required = [random.choice(lowers), random.choice(uppers),
                random.choice(digits), random.choice(symbols)]
    pool = lowers + uppers + digits + symbols
    rest = random.choices(pool, k=length - 4)
    chars = required + rest
    random.shuffle(chars)
    return "".join(chars)

print("OTP:", generate_otp())
print("temp:", generate_temp_password())
print("strong:", generate_strong_password())
```
