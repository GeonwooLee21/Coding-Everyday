## 1️⃣ 파일 읽기 / Reading Files

파일을 읽는 3가지 기본 방법이 있습니다.
There are three core ways to read a file.

| 방법 / Method | 반환값 / Returns | 언제 쓰나 / When |
|---|---|---|
| `f.read()` | 파일 전체를 **하나의 문자열** / one big string | 파일이 작고 전체가 필요할 때 |
| `f.readlines()` | **줄들의 리스트** / a list of lines | 줄 목록이 필요할 때 |
| `for line in f` | 한 줄씩 **반복** / iterate line by line | 줄 단위로 처리할 때 (가장 흔함) |

> ⚠️ 파일을 열면(`open`) 반드시 닫아야(`close`) 합니다. 잠시 후 `with`로 이 문제를 깔끔하게 해결합니다.
> Always `close()` what you `open()`. We'll fix this cleanly with `with` soon.

### 예제 1-1 · `read()` — 전체를 한 문자열로 / Read the whole file

````python
f = open("diary.txt", "r", encoding="utf-8")
content = f.read()       # 파일 전체가 하나의 문자열로 들어옴
f.close()                # 꼭 닫기!

print(content)
print("타입 / type:", type(content))
````

### 예제 1-2 · `for line in f` — 한 줄씩 반복 / Iterate line by line

각 줄 끝에는 줄바꿈 문자 `\n`이 붙어 있습니다. `.strip()`으로 정리하세요.
Each line ends with a newline `\n`; use `.strip()` to clean it.

````python
f = open("diary.txt", "r", encoding="utf-8")
for line in f:
    print("→", line.strip())   # strip()으로 양끝 공백/줄바꿈 제거
f.close()
````

### 예제 1-3 · `readlines()` — 줄들의 리스트로 / Get a list of lines

### ✍️ [실습 1] Wonder Exchange 환율 로그 읽기 / Read the rate log

**과제:** `wonder_exchange_rates.txt`에는 하루치 USD→KRW 환율이 **한 줄에 하나씩** 들어 있습니다.
파일을 읽어 **평균 환율**을 계산하고 소수점 첫째 자리까지 출력하세요.

**Task:** The file has one daily USD→KRW rate per line. Read it and print the **average rate** (1 decimal place).

먼저 아래 셀을 실행해 데이터를 만든 뒤, 그다음 셀에서 직접 코드를 작성해 보세요.

````python
# 데이터 준비 / Prepare data (run this first)
with open("wonder_exchange_rates.txt", "w", encoding="utf-8") as f:
    f.write("1320\n1335\n1318\n1342\n1329\n")
print("✅ wonder_exchange_rates.txt 준비 완료!")
````

````python
# ✍️ 여기에 직접 작성하세요 / Write your code here
# 힌트: 빈 줄을 건너뛰고, int()로 변환해서 리스트에 모으세요.
exchange_rates = []
with open("wonder_exchange_rates.txt", "r", encoding="utf-8") as f:
    for i, line in enumerate(f, 1): # enumerate()는 (인덱스, 값) 순서로 반환함
        if line.strip():
            exchange_rates.append(int(line.strip()))

avg_exchange_rate = sum(exchange_rates) / len(exchange_rates)
print(f"평균 환율: {avg_exchange_rate:.2f}")
````

**✅ 예시 정답 / Sample Solution**

## 2️⃣ 파일 쓰기 / Writing Files

`"w"`(write) 모드로 파일을 엽니다.
Open a file in `"w"` (write) mode.

- `f.write(text)` — 문자열 하나를 씀 / writes one string
- `f.writelines(list)` — 문자열 리스트를 씀 / writes a list of strings (줄바꿈은 직접 넣어야 함!)

> 🚨 **경고 / Warning:** `"w"` 모드는 **기존 내용을 전부 지우고** 새로 씁니다!
> `"w"` mode **erases everything** in the file first. 기존 내용을 지키려면 다음 단원의 `"a"`를 쓰세요.

### 예제 2-1 · `write()` — 한 줄씩 쓰기 / Write line by line

````python
with open("cafe_orders.txt", "w", encoding="utf-8") as f:
    f.write("아메리카노\n")     # \n을 직접 넣어야 줄이 바뀝니다
    f.write("카페라떼\n")

# 잘 써졌는지 확인 / verify
with open("cafe_orders.txt", "r", encoding="utf-8") as f:
    print(f.read())
````

### 예제 2-2 · `writelines()` — 리스트를 한 번에 / Write a list at once

⚠️ `writelines()`는 줄바꿈을 자동으로 넣어주지 **않습니다**. 직접 `\n`을 포함하세요.
`writelines()` does **not** add newlines for you.

````python
menu = ["바닐라라떼\n", "콜드브루\n", "에스프레소\n"]

with open("cafe_orders.txt", "w", encoding="utf-8") as f:   # ⚠️ "w" → 기존 내용 덮어씀!
    f.writelines(menu)

with open("cafe_orders.txt", "r", encoding="utf-8") as f:
    print(f.read())
# 위 예제 2-1의 아메리카노/카페라떼는 사라졌습니다 — "w"가 덮어썼기 때문!
````

### ✍️ [실습 2] Wonder Pay 영수증 쓰기 / Write a receipt

**과제:** 아래 `items` 리스트를 사용해 `wonder_pay_receipt.txt`에 영수증을 작성하세요.
- 첫 줄: `=== Wonder Pay 영수증 ===`
- 각 항목: `상품명: 가격원`
- 마지막 줄: `합계: 총액원`

**Task:** Write a receipt: a header line, one line per item (`name: price원`), and a total line.

````python
# ✍️ 여기에 직접 작성하세요 / Write your code here
items = [("아메리카노", 4500), ("샌드위치", 6000)]

with open("cafe_orders.txt", "w", encoding="utf-8") as f:
    f.write("=== Wonder Pay 영수증===\n")
    total_price = 0
    for item in items:
        name, price = item
        total_price += price
        f.write(f"{name}: {price}원\n")
    f.write(f"합계: {total_price}원")
````

**✅ 예시 정답 / Sample Solution**

## 3️⃣ 파일에 추가하기 / Appending to Files

`"a"`(append) 모드는 기존 내용을 **지우지 않고**, 파일 **끝에 이어서** 씁니다.
`"a"` (append) mode keeps existing content and adds to the **end**.

> 💡 파일이 없으면 `"a"`와 `"w"` 모두 새로 만들어 줍니다. 차이는 **기존 내용 보존 여부**입니다.
> If the file doesn't exist, both create it. The difference is whether existing content is **kept**.

### 예제 3-1 · 일기 한 줄 추가 / Append a line to the diary

````python
# 추가 전 줄 수 / lines before
with open("diary.txt", "r", encoding="utf-8") as f:
    print("추가 전 / before:", len(f.readlines()), "줄")

# 끝에 한 줄 추가 / append one line
with open("diary.txt", "a", encoding="utf-8") as f:
    f.write("파일 추가(append) 모드도 배웠다.\n")

with open("diary.txt", "r", encoding="utf-8") as f:
    print("추가 후 / after:", len(f.readlines()), "줄")

with open("diary.txt", "r", encoding="utf-8") as f:
    print("---\n" + f.read())
````

## 4️⃣ `with` 구문 — 안전한 파일 다루기 / The `with` Statement

지금까지 계속 `with`를 써 왔습니다. 왜 좋을까요?
We've been using `with` all along. Why is it better?

```python
# ❌ 옛날 방식 / old way — close()를 잊거나, 에러 나면 파일이 안 닫힘
f = open("data.txt", "r", encoding="utf-8")
content = f.read()
f.close()   # 이 줄 전에 에러가 나면? 파일이 영영 안 닫힘!

# ✅ with 방식 / with way — 블록을 벗어나면 자동으로 close() (에러가 나도!)
with open("data.txt", "r", encoding="utf-8") as f:
    content = f.read()
# 여기서 파일은 이미 닫혀 있음
```

핵심: `with` 블록을 빠져나가는 순간 — **정상 종료든 에러든** — 파일이 자동으로 닫힙니다.
Key idea: the file is closed automatically when the block ends, **even if an error occurs**.

### 예제 4-1 · `with`가 정말 자동으로 닫는지 확인 / Proof that `with` auto-closes

````python
with open("diary.txt", "r", encoding="utf-8") as f:
    first_line = f.readline().strip()
    print("블록 안 / inside  — f.closed =", f.closed)   # False

print("블록 밖 / outside — f.closed =", f.closed)        # True (자동으로 닫힘!)
print("첫 줄 / first line:", first_line)
````

### ✍️ [실습 4] 읽고 → 가공해서 → 새 파일에 쓰기 / Read, transform, write

**과제:** `wonder_exchange_rates.txt`(실습 1에서 생성)를 `with`로 읽어서,
각 환율에 **+5원** 우대 환율을 적용한 값을 `preferred_rates.txt`에 한 줄씩 쓰세요.

**Task:** Read the rate file with `with`, add 5 to each rate, and write the results to a new file — one per line.

````python
# ✍️ 여기에 직접 작성하세요 / Write your code here
# 힌트: 읽기 with 블록과 쓰기 with 블록을 따로 쓰면 깔끔합니다.
exchange_rates = []
with open("wonder_exchange_rates.txt", "r", encoding="utf-8") as f:
    for line in f:
        if line.strip("\n"):
            exchange_rates.append(int(line.strip("\n")))

with open("preferred_rates.txt", "w", encoding="utf-8") as f:
    for rate in exchange_rates:
        f.write(f"{rate+5}\n")
````

**✅ 예시 정답 / Sample Solution**

````python
# 1) 읽기 / read
with open("wonder_exchange_rates.txt", "r", encoding="utf-8") as f:
    rates = [int(line.strip()) for line in f if line.strip()]

# 2) 가공 / transform
preferred = [rate + 5 for rate in rates]

# 3) 쓰기 / write
with open("preferred_rates.txt", "w", encoding="utf-8") as f:
    for rate in preferred:
        f.write(f"{rate}\n")

# 확인 / verify
with open("preferred_rates.txt", "r", encoding="utf-8") as f:
    print(f.read())
````

## 5️⃣ CSV 파일 다루기 / Working with CSV

CSV(Comma-Separated Values)는 **표 형태 데이터**를 줄·쉼표로 저장하는 형식입니다.
`csv` 모듈을 쓰면 쉼표 분리, 따옴표 처리를 자동으로 해 줍니다 — 직접 `split(",")` 하지 마세요!

CSV stores **table-like data**. The `csv` module handles comma-splitting and quoting for you.

> 💡 CSV 파일을 열 때는 `newline=""`을 함께 지정합니다 (빈 줄이 끼는 문제 방지).
> When opening CSV files, add `newline=""` to avoid blank-line issues.

### 예제 5-1 · `csv.writer` — 쓰기 / Writing rows

````python
import csv

rows = [
    ["name", "amount"],   # 헤더 / header row
    ["민준", 12000],
    ["서연", 4500],
    ["지호", 30000],
]

with open("payments.csv", "w", encoding="utf-8", newline="") as f:
    writer = csv.writer(f)
    writer.writerows(rows)     # 리스트들을 한 번에 / write all rows

print("✅ payments.csv 작성 완료!")
````

### 예제 5-2 · `csv.reader` — 읽기 / Reading rows

`csv.reader`는 각 줄을 **문자열 리스트**로 줍니다. 숫자는 직접 `int()` 변환해야 합니다.
`csv.reader` gives each row as a **list of strings** — convert numbers yourself.

````python
import csv

with open("payments.csv", "r", encoding="utf-8", newline="") as f:
    reader = csv.reader(f)
    header = next(reader)        # 첫 줄(헤더)을 따로 빼기 / pull off the header
    print("헤더 / header:", header)
    for row in reader:
        name, amount = row
        print(f"{name} → {int(amount):,}원")
````

### 예제 5-3 · `csv.DictReader` — 헤더를 key로 / Read rows as dictionaries

헤더 이름으로 값에 접근할 수 있어 가독성이 좋습니다.
Access values by column name — more readable.

````python
import csv

with open("payments.csv", "r", encoding="utf-8", newline="") as f:
    reader = csv.DictReader(f)
    total = 0
    for record in reader:
        total += int(record["amount"])   # 헤더 이름으로 접근
print(f"총 결제액 / total: {total:,}원")
````

### ✍️ [실습 5] Wonder Pay 일일 정산 / Daily settlement

**과제:**
1. 아래 거래 데이터를 `daily_payments.csv`에 헤더(`name,amount`)와 함께 쓰세요.
2. 그 파일을 다시 읽어 **총 결제액**과 **거래 건수**를 출력하세요.

**Task:** Write the data to a CSV with a header, then read it back and print the total amount and number of transactions.

````python
# ✍️ 여기에 직접 작성하세요 / Write your code here
import csv
transactions = [("하은", 8000), ("도윤", 15000), ("수아", 6500), ("이안", 22000)]

with open("daily_payments.csv", "w", encoding="utf-8") as f:
    f.write("name,amount\n")
    writer = csv.writer(f)
    writer.writerows(transactions)

with open("daily_payments.csv", "r", encoding="utf-8") as f:
    reader = csv.DictReader(f)
    total = 0
    count = 0
    for data in reader:
        total += int(data["amount"])
        count += 1
print(f"총 결제액: {total}, 거래 건수: {count}")
````

**✅ 예시 정답 / Sample Solution**

````python
import csv
transactions = [("하은", 8000), ("도윤", 15000), ("수아", 6500), ("이안", 22000)]

# 1) 쓰기 / write
with open("daily_payments.csv", "w", encoding="utf-8", newline="") as f:
    writer = csv.writer(f)
    writer.writerow(["name", "amount"])      # 헤더
    writer.writerows(transactions)           # 데이터

# 2) 읽어서 집계 / read back and aggregate
with open("daily_payments.csv", "r", encoding="utf-8", newline="") as f:
    reader = csv.reader(f)
    next(reader)                             # 헤더 건너뛰기
    total = 0
    count = 0
    for name, amount in reader:
        total += int(amount)
        count += 1

print(f"거래 건수 / transactions: {count}건")
print(f"총 결제액 / total amount: {total:,}원")
````

`spending.txt`에는 결제 금액이 **한 줄에 하나씩** (정수) 들어 있습니다.
파일을 읽어 **모든 금액의 합**을 반환하는 함수를 작성하세요. **빈 줄은 무시**해야 합니다.

Read a file of integer amounts (one per line) and return their **sum**. **Ignore blank lines.**

### 📝 퀴즈 3 — 고액 결제 건수 / Count Large Payments  *(CSV)*

CSV 파일(헤더: `name,amount`)을 읽어, **금액이 `threshold`를 초과(>)하는** 결제 건수를 반환하세요.
헤더 줄은 세면 안 됩니다.

Read a CSV with header `name,amount` and return how many rows have `amount` **strictly greater than** `threshold`.

```python
import csv

    # 여기에 작성 / your code here
    pass
```

````python
# 퀴즈 3 — 여기에 함수를 작성하세요 / Write your function here
import csv
def count_large_payments(filename: str, threshold: int) -> int:
    with open(f"{filename}", "r", encoding="utf-8") as f:
        reader = csv.DictReader(f)
        count = 0
        for data in reader:
            if int(data["amount"]) > threshold:
                count += 1
    return count

# --- 테스트 / tests ---
with open("quiz3_test.csv", "w", encoding="utf-8", newline="") as f:
    w = csv.writer(f)
    w.writerow(["name", "amount"])
    w.writerow(["민준", 12000])
    w.writerow(["서연", 4500])
    w.writerow(["지호", 30000])
    w.writerow(["하은", 10000])

print("결과 / result:", count_large_payments("quiz3_test.csv", 10000))
print("기대값 / expected: 2  (12000, 30000 — 10000은 초과가 아님)")
````
