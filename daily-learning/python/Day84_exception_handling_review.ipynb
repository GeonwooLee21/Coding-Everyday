# 1️⃣ `try` & `except` — 에러 잡기 / Catching Errors

## 🤔 왜 필요할까? / Why do we need it?

먼저 예외 처리가 **없는** 코드를 봅시다. 아래 셀을 실행하면 프로그램이 멈춥니다 (에러 메시지가 빨간색으로 나타나요).

First, let's look at code **without** exception handling. Running the cell below crashes the program (you'll see a red error message).

````python
# ❌ 예외 처리가 없는 코드 / Code WITHOUT exception handling
# 이 셀은 일부러 에러를 발생시킵니다! / This cell raises an error on purpose!

user_input = "오천원"          # 사용자가 숫자 대신 한글을 입력함 / User typed text instead of a number
amount = int(user_input)       # 💥 여기서 ValueError 발생! / ValueError happens here!
print(f"결제 금액: {amount}원")  # 이 줄은 실행되지 않음 / This line never runs
````

## 🛡️ `try` & `except` 기본 구조 / Basic Structure

```python
try:
    # 에러가 발생할 수 있는 코드 / Code that might cause an error
except:
    # 에러가 발생했을 때 실행되는 코드 / Code that runs when an error happens
```

- `try` 블록 안에서 에러가 발생하면, 그 즉시 `except` 블록으로 이동합니다.
- 에러가 없으면 `except` 블록은 건너뜁니다.

- If an error occurs inside the `try` block, Python immediately jumps to the `except` block.
- If there is no error, the `except` block is skipped.

````python
# ✅ 예외 처리가 있는 코드 / Code WITH exception handling

user_input = "오천원"

try:
    amount = int(user_input)
    print(f"결제 금액: {amount}원")
except:
    print("⚠️ 숫자를 입력해주세요! / Please enter a number!")

print("프로그램이 멈추지 않고 계속 실행됩니다! / The program keeps running!")
````

## 🎯 특정 예외만 잡기 / Catching Specific Exceptions

`except:`만 쓰면 **모든** 에러를 잡아버립니다. 이러면 진짜 버그도 숨겨버려서 위험해요! 
**어떤 에러를 잡을지 명시하는 것**이 좋은 습관입니다.

A bare `except:` catches **every** error — including real bugs you'd want to see! 
It's a good habit to **specify which error you want to catch**.

```python
try:
    ...
except ValueError:        # ValueError만 잡음 / Catches only ValueError
    ...
```

`as` 키워드를 쓰면 에러 메시지도 확인할 수 있습니다. / With the `as` keyword, you can inspect the error message.

````python
# 특정 예외 잡기 + 에러 메시지 확인하기
# Catching a specific exception + inspecting the message

user_input = "5,000"   # 쉼표 때문에 int() 변환 실패 / int() fails because of the comma

try:
    amount = int(user_input)
except ValueError as error:
    print("⚠️ 변환 실패! / Conversion failed!")
    print(f"에러 내용 / Error details: {error}")
````

## 🎯 여러 종류의 예외 잡기 / Catching Multiple Exception Types

`except` 블록을 여러 개 쓰면, 에러 종류에 따라 다르게 대응할 수 있습니다.

With multiple `except` blocks, you can respond differently depending on the error type.

````python
# 더치페이 계산기 / Bill-splitting calculator
# people 값을 0, "둘", 4로 바꿔가며 실행해보세요!
# Try changing people to 0, "둘", and 4!

total = 30000
people = 0

try:
    share = total / people
    print(f"1인당 금액: {share}원 / Per person: {share} KRW")
except ZeroDivisionError:
    print("⚠️ 0명으로 나눌 수 없어요! / Cannot divide by zero people!")
except TypeError:
    print("⚠️ 인원 수는 숫자여야 해요! / People count must be a number!")
````

## 🧪 Lab 1-1: 안전한 숫자 변환기 / Safe Number Converter

Wonder Pay에서 사용자가 입력한 금액 문자열을 정수로 바꾸는 함수를 만들어 봅시다.

Let's build a function that converts a user's amount string into an integer for Wonder Pay.

**요구사항 / Requirements:**
- 변환에 성공하면 그 정수를 반환 / If conversion succeeds, return the integer
- `ValueError`가 발생하면 `None`을 반환 / If a `ValueError` occurs, return `None`

````python
def safe_to_int(text):
    # TODO 1: try 블록 안에서 text를 int로 변환하고 반환하세요
    # TODO 1: Inside a try block, convert text to int and return it
    try:
        number = int(text)
        return number
    # TODO 2: ValueError를 잡아서 None을 반환하세요
    # TODO 2: Catch ValueError and return None
    except ValueError:
        return None


# ===== 테스트 / Tests (수정 금지 / Do not modify) =====
print(safe_to_int("1000"))    # 예상 / Expected: 1000
print(safe_to_int("-50"))     # 예상 / Expected: -50
print(safe_to_int("hello"))   # 예상 / Expected: None
print(safe_to_int("12.5"))    # 예상 / Expected: None (int()는 "12.5"를 변환하지 못해요!)

if safe_to_int("1000") == 1000 and safe_to_int("hello") is None:
    print("✅ Lab 1-1 통과! / Passed!")
else:
    print("❌ 다시 확인해보세요 / Check again")
````

## 🧪 Lab 1-2: 더치페이 함수 만들기 / Bill-Splitting Function

위에서 본 더치페이 계산을 함수로 만들어 봅시다. 에러 종류에 따라 다른 메시지를 출력해야 합니다.

Let's turn the bill-splitting logic into a function. It must print different messages depending on the error type.

**요구사항 / Requirements:**
- 성공하면 1인당 금액 반환 / On success, return the per-person share
- `ZeroDivisionError` 또는 `TypeError`가 발생하면 경고 출력 후 `0` 반환 / On `ZeroDivisionError` or `TypeError`, print a warning and return `0`

````python
def split_bill(total, people):
    try:
        share = total / people
        return share
    # TODO 1: 0으로 나눌 때 발생하는 예외를 잡으세요
    # TODO 1: Catch the exception raised when dividing by zero
    except ZeroDivisionError:
        print("⚠️ 인원 수는 0이 될 수 없습니다! / People count cannot be zero!")
        return 0
    # TODO 2: 숫자가 아닌 값으로 나눌 때 발생하는 예외를 잡으세요
    # TODO 2: Catch the exception raised when dividing by a non-number
    except TypeError:
        print("⚠️ 숫자만 입력하세요! / Numbers only, please!")
        return 0


# ===== 테스트 / Tests (수정 금지 / Do not modify) =====
print(split_bill(30000, 3))       # 예상 / Expected: 10000.0
print(split_bill(30000, 0))       # 예상 / Expected: 경고 후 0 / warning then 0
print(split_bill(30000, "셋"))    # 예상 / Expected: 경고 후 0 / warning then 0

if split_bill(30000, 3) == 10000 and split_bill(30000, 0) == 0:
    print("✅ Lab 1-2 통과! / Passed!")
else:
    print("❌ 다시 확인해보세요 / Check again")
````

# 2️⃣ `else` — 에러가 없을 때만 / Only When There's No Error

`try` 블록에서 **에러가 발생하지 않았을 때만** 실행되는 블록입니다.

The `else` block runs **only if no error occurred** in the `try` block.

```python
try:
    # 에러가 발생할 수 있는 코드 / Code that might fail
except SomeError:
    # 에러 발생 시 / If an error happens
else:
    # 에러가 없을 때만 실행 / Runs only if NO error happened
```

🤔 **"그냥 try 안에 다 넣으면 안 되나요?"** / *"Can't we just put everything inside `try`?"*

넣을 수는 있지만, `else`를 쓰면 **"위험한 코드"와 "성공했을 때의 코드"가 명확하게 분리**됩니다. `try` 블록은 짧을수록 좋아요 — 어디서 에러가 날 수 있는지 한눈에 보이니까요!

You can, but `else` clearly separates the *risky code* from the *success code*. The shorter the `try` block, the easier it is to see exactly where an error can come from!

````python
# Wonder Exchange 환율 조회 / Wonder Exchange rate lookup
# currency 값을 "USD", "GBP"로 바꿔가며 실행해보세요!
# Try changing currency to "USD" and "GBP"!

exchange_rates = {"USD": 1350.50, "EUR": 1480.20, "JPY": 9.15}
currency = "USD"

try:
    rate = exchange_rates[currency]   # 위험한 코드 / Risky code (KeyError 가능)
except KeyError:
    print(f"⚠️ '{currency}'은(는) 지원하지 않는 통화입니다. / Unsupported currency.")
else:
    print(f"✅ 1 {currency} = {rate} KRW")   # 성공했을 때만 / Only on success
````

# 3️⃣ `finally` — 무슨 일이 있어도 항상 / No Matter What, Always

`finally` 블록은 에러가 발생하든 안 하든 **항상** 실행됩니다.

The `finally` block **always** runs — whether an error occurred or not.

```python
try:
    # 위험한 코드 / Risky code
except SomeError:
    # 에러 발생 시 / On error
finally:
    # 항상 실행! / ALWAYS runs!
```

💡 **언제 쓸까요? / When do we use it?** 
파일 닫기, 네트워크 연결 종료처럼 **"뒷정리"가 반드시 필요한 작업**에 사용합니다. 에러가 나서 중간에 멈추더라도 뒷정리는 해야 하니까요!

It's used for **clean-up work that must happen** — like closing files or network connections — even if an error interrupted the work in the middle!

````python
# finally의 동작 확인하기 / Watching finally in action
# value 값을 10, 0으로 바꿔가며 실행해보세요!
# Try changing value to 10 and 0!

value = 10

try:
    result = 100 / value
    print(f"결과 / Result: {result}")
except ZeroDivisionError:
    print("⚠️ 0으로 나눌 수 없어요! / Cannot divide by zero!")
finally:
    print("🧹 뒷정리 완료! 이 줄은 항상 출력됩니다. / Clean-up done! This always prints.")
````

## 🧪 Lab 3-1: 거래 내역 파일 안전하게 읽기 / Safely Reading a Transaction File

Wonder Pay의 거래 내역 파일을 읽는 함수를 만들어 봅시다. 파일이 없어도 프로그램이 멈추면 안 되고, **파일을 열었다면 반드시 닫아야** 합니다.

Let's build a function that reads Wonder Pay's transaction file. The program must not crash if the file is missing, and **if a file was opened, it must be closed**.

**요구사항 / Requirements:**
- 파일이 없으면(`FileNotFoundError`) 경고 출력 후 빈 문자열 `""` 반환
- `finally` 블록에서 파일이 열려 있으면 닫기
- If the file is missing (`FileNotFoundError`), print a warning and return `""`
- In the `finally` block, close the file if it was opened

````python
# 먼저 연습용 파일을 만듭니다 / First, create a practice file
with open("transactions.txt", "w", encoding="utf-8") as f:
    f.write("WonderPay,5000")
print("연습용 파일 생성 완료! / Practice file created!")
````

````python
def read_transaction_file(filename):
    file = None
    try:
        file = open(filename, "r", encoding="utf-8")
        content = file.read()
        return content
    # TODO 1: 파일이 없을 때 발생하는 예외를 잡으세요
    # TODO 1: Catch the exception raised when the file does not exist
    except FileNotFoundError:
        print(f"⚠️ 파일을 찾을 수 없습니다: {filename}")
        return ""
    # TODO 2: 항상 실행되는 블록을 완성하세요
    # TODO 2: Complete the block that always runs
    finally:
        if file is not None:
            file.close()
            print("📁 파일을 닫았습니다. / File closed.")


# ===== 테스트 / Tests (수정 금지 / Do not modify) =====
print(read_transaction_file("transactions.txt"))    # 예상: 내용 출력 + 파일 닫힘 메시지
print(read_transaction_file("ghost_file.txt"))      # 예상: 경고 출력 (멈추지 않음!)

if read_transaction_file("transactions.txt") == "WonderPay,5000":
    print("✅ Lab 3-1 통과! / Passed!")
else:
    print("❌ 다시 확인해보세요 / Check again")
````

## 📐 전체 구조 정리 / The Full Structure

네 가지를 모두 합치면 이런 모양이 됩니다. / Putting all four together:

```python
try:
    # 1. 위험한 코드 / Risky code
except SomeError:
    # 2. 에러 발생 시 / If an error happened
else:
    # 3. 에러가 없을 때만 / Only if NO error
finally:
    # 4. 항상! / ALWAYS!
```

| 블록 / Block | 실행 조건 / When does it run? |
|---|---|
| `try` | 항상 시도 / Always attempted |
| `except` | 에러가 발생했을 때만 / Only when an error occurs |
| `else` | 에러가 **없을** 때만 / Only when there is **no** error |
| `finally` | **항상** / **Always** |

````python
# 네 블록이 모두 있는 예제 / All four blocks together
# amount_text를 "5000", "오천원"으로 바꿔가며 실행해보세요!
# Try changing amount_text to "5000" and "오천원"!

amount_text = "5000"

try:
    amount = int(amount_text)
except ValueError:
    print("⚠️ [except] 숫자가 아니에요! / Not a number!")
else:
    print(f"✅ [else] 결제 금액 확인: {amount}원 / Payment confirmed: {amount} KRW")
finally:
    print("🧾 [finally] 결제 시도가 기록되었습니다. / Payment attempt logged.")
````

# 4️⃣ `raise` — 내가 직접 에러 발생시키기 / Raising Errors Yourself

지금까지는 Python이 발생시킨 에러를 **잡기만** 했습니다. 그런데 때로는 **우리가 직접** 에러를 발생시켜야 할 때가 있어요!

So far we've only been **catching** errors that Python raised. But sometimes **we** need to raise errors ourselves!

```python
raise ValueError("에러 메시지 / Error message")
```

🤔 **왜 일부러 에러를 만들까요? / Why would we cause errors on purpose?**

Wonder Pay에서 누군가 **-5000원을 입금**하려고 한다고 생각해보세요. Python 입장에서 `-5000`은 멀쩡한 정수라서 에러가 나지 않아요. 하지만 **비즈니스 규칙**으로는 명백한 잘못이죠! 이럴 때 `raise`로 "이건 잘못된 상황이야!"라고 알리는 겁니다.

Imagine someone tries to **deposit -5,000 KRW** in Wonder Pay. To Python, `-5000` is a perfectly fine integer — no error. But by **business rules**, it's clearly wrong! `raise` lets us shout "this situation is invalid!"

````python
# raise 기본 사용법 / Basic raise usage
# 이 셀은 일부러 에러를 발생시킵니다! / This cell raises an error on purpose!

deposit = -5000

if deposit <= 0:
    raise ValueError(f"입금액은 0보다 커야 합니다. 입력값: {deposit} / Deposit must be positive. Got: {deposit}")

print("이 줄은 실행되지 않습니다 / This line never runs")
````

💡 **중요한 연결고리 / The important connection:** 
`raise`로 발생시킨 에러도 평범한 에러와 똑같이 `try` / `except`로 잡을 수 있습니다!

Errors raised with `raise` can be caught with `try` / `except`, just like any other error!

````python
# raise한 에러를 except로 잡기 / Catching an error we raised

def check_deposit(amount):
    if amount <= 0:
        raise ValueError("입금액은 0보다 커야 합니다. / Deposit must be positive.")
    print(f"✅ {amount}원 입금 가능 / Deposit OK")

try:
    check_deposit(-5000)
except ValueError as error:
    print(f"⚠️ 입금 거부됨 / Deposit rejected: {error}")
````

## 🧪 Lab 4-1: 입금 검증 함수 / Deposit Validator

Wonder Pay의 입금 검증 함수를 만들어 봅시다. / Let's build Wonder Pay's deposit validator.

**요구사항 / Requirements:**
- `amount`가 정수가 아니면 → `TypeError` 발생 / If `amount` is not an integer → raise `TypeError`
- `amount`가 0 이하이면 → `ValueError` 발생 / If `amount` is 0 or less → raise `ValueError`
- 둘 다 통과하면 → `True` 반환 / If both checks pass → return `True`

💡 힌트 / Hint: 정수인지 확인할 때는 `isinstance(amount, int)`를 사용하세요. / Use `isinstance(amount, int)` to check for an integer.

````python
def validate_deposit(amount):
    # TODO 1: amount가 int가 아니면 TypeError를 발생시키세요
    # TODO 1: If amount is not an int, raise TypeError
    if not isinstance(amount, int):
        raise TypeError("입금액은 정수여야 합니다. / Deposit must be an integer.")
    # TODO 2: amount가 0 이하이면 ValueError를 발생시키세요
    # TODO 2: If amount is 0 or less, raise ValueError
    if amount <= 0:
        raise ValueError("입금액은 0보다 커야 합니다. / Deposit must be positive.")
    return True


# ===== 테스트 / Tests (수정 금지 / Do not modify) =====
test_results = []

try:
    test_results.append(validate_deposit(1000) is True)
except (ValueError, TypeError):
    test_results.append(False)

try:
    validate_deposit(-500)
    test_results.append(False)            # 에러가 안 나면 실패 / Fail if no error
except ValueError:
    test_results.append(True)             # ValueError가 나야 정답 / ValueError = correct

try:
    validate_deposit("1000")
    test_results.append(False)
except TypeError:
    test_results.append(True)

if all(test_results):
    print("✅ Lab 4-1 통과! / Passed!")
else:
    print("❌ 다시 확인해보세요 / Check again:", test_results)
````

## 🧪 Lab 4-2: Wonder Pay 계좌 클래스 / Wonder Pay Account Class

OOP에서 배운 클래스와 `raise`를 결합해 봅시다! / Let's combine the classes you learned in OOP with `raise`!

**요구사항 / Requirements:**
- `withdraw(amount)` 메서드에서:
  - `amount`가 0 이하 → `ValueError` 발생 / `amount` ≤ 0 → raise `ValueError`
  - `amount`가 잔액보다 크면 → `ValueError` 발생 / `amount` > balance → raise `ValueError`
  - 정상이면 잔액을 차감하고 새 잔액 반환 / Otherwise, deduct and return the new balance
- ⚠️ **주의:** 출금이 실패하면 잔액이 **변하면 안 됩니다!** / If a withdrawal fails, the balance **must not change!**

````python
class WonderPayAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self.balance = balance

    def withdraw(self, amount):
        # TODO 1: amount가 0 이하이면 ValueError를 발생시키세요
        # TODO 1: If amount is 0 or less, raise ValueError
        if amount <= 0:
            raise ValueError("출금액은 0보다 커야 합니다. / Withdrawal must be positive.")
        # TODO 2: amount가 잔액(self.balance)보다 크면 ValueError를 발생시키세요
        # TODO 2: If amount exceeds the balance, raise ValueError
        if amount > self.balance:
            raise ValueError(f"잔액 부족! 현재 잔액: {self.balance}원 / Insufficient balance")
        # TODO 3: 잔액을 차감하고 새 잔액을 반환하세요
        # TODO 3: Deduct the amount and return the new balance
        self.balance -= amount
        return self.balance


# ===== 테스트 / Tests (수정 금지 / Do not modify) =====
account = WonderPayAccount("민지", 10000)
print(f"출금 후 잔액 / Balance after withdrawal: {account.withdraw(3000)}원")   # 예상: 7000

try:
    account.withdraw(50000)
    print("❌ 에러가 발생해야 합니다! / An error should have occurred!")
except ValueError as error:
    print(f"⚠️ 출금 거부 / Withdrawal rejected: {error}")

if account.balance == 7000:
    print("✅ Lab 4-2 통과! 실패한 출금은 잔액을 바꾸지 않았어요. / Passed! Failed withdrawal didn't change the balance.")
else:
    print(f"❌ 잔액이 {account.balance}원입니다. 7000원이어야 해요! / Balance should be 7000!")
````

## 🧪 Lab 5-2: 통합 실습 — 결제 일괄 처리기 / Integrated Lab — Batch Payment Processor

Wonder Pay 서버에 하루치 결제 요청이 도착했습니다. 그런데 데이터가 엉망이에요! 문자열 숫자, 한글, `None`, 음수가 섞여 있습니다. **하나라도 이상하다고 전체 처리가 멈추면 안 됩니다.**

A day's worth of payment requests just arrived at the Wonder Pay server — and the data is a mess! String numbers, Korean text, `None`, and negatives are all mixed in. **One bad item must not stop the whole batch.**

**요구사항 / Requirements:**
1. 각 항목을 `int()`로 변환 시도 / Try converting each item with `int()`
2. 변환된 금액이 0 이하이면 직접 `ValueError`를 발생시키기 / If the amount is ≤ 0, raise `ValueError` yourself
3. `ValueError`와 `TypeError`를 잡아서 실패 횟수만 세고 계속 진행 / Catch `ValueError` and `TypeError`, count the failure, and keep going
4. (성공 건수, 실패 건수, 총 결제액) 반환 / Return (success count, fail count, total amount)

💡 힌트 / Hint: 두 예외를 한 번에 잡으려면 `except (ValueError, TypeError):` 처럼 튜플을 사용하세요! / To catch two exceptions at once, use a tuple: `except (ValueError, TypeError):`

````python
def process_payments(payments):
    success_count = 0
    fail_count = 0
    total = 0
    for item in payments:
        try:
            # TODO 1: item을 int로 변환하세요 / Convert item to int
            amount = int(item)
            # TODO 2: amount가 0 이하이면 ValueError를 발생시키세요
            # TODO 2: If amount is 0 or less, raise ValueError
            if amount <= 0:
                raise ValueError("금액은 양수여야 합니다. / Amount must be positive.")
            total += amount
            success_count += 1
        # TODO 3: ValueError와 TypeError를 한 번에 잡으세요
        # TODO 3: Catch ValueError and TypeError together
        except (ValueError, TypeError):
            fail_count += 1
    return success_count, fail_count, total


# ===== 테스트 / Tests (수정 금지 / Do not modify) =====
daily_payments = ["5000", "3000", "오천원", "-100", None, "2000"]
success, fail, total = process_payments(daily_payments)
print(f"성공 / Success: {success}건, 실패 / Fail: {fail}건, 총액 / Total: {total}원")

if (success, fail, total) == (3, 3, 10000):
    print("✅ Lab 5-2 통과! 서버가 멈추지 않았어요! / Passed! The server never crashed!")
else:
    print("❌ 예상: (3, 3, 10000) / Expected: (3, 3, 10000)")
````

````python
def split_bill(total, people):
    try:
        share = total / people
        return share
    except ZeroDivisionError:
        print("⚠️ 인원 수는 0이 될 수 없습니다! / People count cannot be zero!")
        return 0
    except TypeError:
        print("⚠️ 숫자만 입력하세요! / Numbers only, please!")
        return 0

# 💡 해설 / Explanation:
# except 블록은 위에서부터 차례로 검사됩니다. 발생한 에러와 일치하는 첫 번째 블록만 실행돼요.
# except blocks are checked top to bottom. Only the FIRST matching block runs.
print(split_bill(30000, 3))
````

## 🔑 Lab 3-1 정답 / Solution

````python
def read_transaction_file(filename):
    file = None
    try:
        file = open(filename, "r", encoding="utf-8")
        content = file.read()
        return content
    except FileNotFoundError:
        print(f"⚠️ 파일을 찾을 수 없습니다: {filename}")
        return ""
    finally:
        if file is not None:
            file.close()
            print("📁 파일을 닫았습니다. / File closed.")

# 💡 해설 / Explanation:
# 1) return문이 있어도 finally는 실행됩니다! return 직전에 finally가 끼어들어요.
#    finally runs even when there's a return! It sneaks in right before the return happens.
# 2) file = None으로 시작하는 이유: open() 자체가 실패하면 file 변수가 없어서
#    finally에서 NameError가 날 수 있기 때문이에요.
#    We start with file = None because if open() itself fails, the file variable
#    wouldn't exist and finally could raise a NameError.
print(read_transaction_file("transactions.txt"))
````

## 🔑 Lab 4-1 정답 / Solution

````python
def validate_deposit(amount):
    if not isinstance(amount, int):
        raise TypeError("입금액은 정수여야 합니다. / Deposit must be an integer.")
    if amount <= 0:
        raise ValueError("입금액은 0보다 커야 합니다. / Deposit must be positive.")
    return True

# 💡 해설 / Explanation:
# 타입 검사를 먼저 하는 이유: "1000" <= 0 같은 비교는 TypeError를 일으키기 때문에,
# 타입을 먼저 확인해야 의도한 에러 메시지를 전달할 수 있어요.
# We check the type FIRST because a comparison like "1000" <= 0 would itself
# raise a TypeError — checking type first lets us control the error message.
print(validate_deposit(1000))
````

## 🔑 Lab 4-2 정답 / Solution

````python
class WonderPayAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self.balance = balance

    def withdraw(self, amount):
        if amount <= 0:
            raise ValueError("출금액은 0보다 커야 합니다. / Withdrawal must be positive.")
        if amount > self.balance:
            raise ValueError(f"잔액 부족! 현재 잔액: {self.balance}원 / Insufficient balance")
        self.balance -= amount
        return self.balance

# 💡 해설 / Explanation:
# 검증(raise)을 모두 통과한 후에만 self.balance를 수정합니다.
# 만약 차감을 먼저 하고 검증하면, 실패한 출금이 잔액을 망가뜨려요! (실제 금융 사고 패턴!)
# We modify self.balance only AFTER all validations pass. If we deducted first
# and validated later, a failed withdrawal would corrupt the balance — a real
# pattern behind actual financial system bugs!
account = WonderPayAccount("민지", 10000)
print(account.withdraw(3000))
````

## 🔑 Lab 5-2 정답 / Solution

````python
def process_payments(payments):
    success_count = 0
    fail_count = 0
    total = 0
    for item in payments:
        try:
            amount = int(item)
            if amount <= 0:
                raise ValueError("금액은 양수여야 합니다. / Amount must be positive.")
            total += amount
            success_count += 1
        except (ValueError, TypeError):
            fail_count += 1
    return success_count, fail_count, total

# 💡 해설 / Explanation:
# - int("오천원") → ValueError (값이 문제 / bad value)
# - int(None)    → TypeError  (타입이 문제 / bad type)
# - int("-100")은 성공하지만, 우리가 직접 raise한 ValueError에 걸립니다!
#   int("-100") succeeds, but it gets caught by the ValueError WE raised!
# - try가 for 루프 "안에" 있어서, 한 항목이 실패해도 다음 항목으로 계속 진행합니다.
#   The try is INSIDE the loop, so one failure doesn't stop the next item.
print(process_payments(["5000", "3000", "오천원", "-100", None, "2000"]))
````

## 🔑 Quiz 2 정답 / Solution

````python
def krw_to_usd(krw_amount, rate):
    if rate <= 0:
        raise ValueError("환율은 0보다 커야 합니다. / Rate must be positive.")
    return krw_amount / rate


def safe_exchange(krw_amount, rate):
    try:
        usd = krw_to_usd(krw_amount, rate)
    except ValueError:
        return -1
    except TypeError:
        return -1
    else:
        return round(usd, 2)

# 💡 해설 / Explanation:
# - rate가 "환율"(문자열)이면 rate <= 0 비교 자체가 TypeError를 발생시킵니다.
#   우리가 raise하지 않은 에러도 except로 잡을 수 있다는 점이 포인트!
#   When rate is a string, the comparison rate <= 0 itself raises TypeError.
#   The point: except catches errors we did NOT raise ourselves, too!
# - except (ValueError, TypeError): return -1 처럼 한 줄로 합쳐도 정답입니다.
#   Combining into one line with a tuple is also a correct answer.
checks = [
    safe_exchange(135050, 1350.50) == 100.0,
    safe_exchange(135050, 0) == -1,
    safe_exchange(135050, -10) == -1,
    safe_exchange(135050, "환율") == -1,
]
print("✅ 모든 케이스 통과!" if all(checks) else "❌", checks)
````

````python
# 🔑 Quiz 3(b) 정답 / Solution

user = {"name": "민지"}

# 방법 1 / Method 1: try & except
try:
    age_1 = user["age"]
except KeyError:
    age_1 = 0

# 방법 2 / Method 2: .get() — 키가 없으면 기본값 반환 / returns a default if the key is missing
age_2 = user.get("age", 0)

# 💡 해설 / Explanation:
# 두 방법 모두 정답이지만 철학이 다릅니다:
# Both are correct, but the philosophies differ:
# - try/except: "일단 시도하고, 실패하면 수습한다" (EAFP 스타일 — Python다운 방식!)
#   "Try first, handle failure" (EAFP style — the Pythonic way!)
# - .get():    "없을 수도 있다는 걸 알고 기본값을 준비한다"
#   "Know it might be missing, prepare a default"
# 단순한 기본값이 필요할 땐 .get()이 더 깔끔하고,
# 실패 시 복잡한 처리(로그 기록, 재시도 등)가 필요할 땐 try/except가 적합해요.
# .get() is cleaner for simple defaults; try/except fits when failure needs
# real handling (logging, retries, etc.).
print(f"age_1 = {age_1}, age_2 = {age_2}")
````
