## 1️⃣ 문자열 메서드 / String Methods

문자열은 **수정 불가능(immutable)**합니다. 모든 문자열 메서드는 원본을 바꾸지 않고 **새 문자열을 반환**합니다.

Strings are **immutable**. Every string method returns a **new string** — the original never changes.

### 1-1. `upper()` / `lower()` / `title()` / `capitalize()` — 대소문자 변환

🇰🇷 문자열의 대소문자를 바꿔주는 메서드입니다. 원본 문자열은 바뀌지 않고, **새 문자열을 반환**합니다. 통화 코드(usd → USD)나 고객 이름 정리에 자주 쓰입니다.

🇺🇸 These methods change the letter case of a string. The original string is unchanged — a **new string is returned**. Useful for normalizing currency codes (usd → USD) or customer names.

| 메서드 Method | 설명 Description |
|---|---|
| `upper()` | 모두 대문자로 / all UPPERCASE |
| `lower()` | 모두 소문자로 / all lowercase |
| `title()` | 각 단어의 첫 글자만 대문자 / Each Word Capitalized |
| `capitalize()` | 첫 글자만 대문자 / First letter only |


```python
# 데모 / Demo — Wonder Exchange 통화 코드 정리
currency_code = "usd"
customer_name = "kim minji"

print(currency_code.upper())       # USD
print("KRW".lower())               # krw
print(customer_name.title())       # Kim Minji
print(customer_name.capitalize())  # Kim minji
```

```python
# 연습 / Practice
# TODO 1: 통화 코드를 대문자로 변환하세요 / Convert the currency code to uppercase
raw_code = "jpy"
clean_code = raw_code.upper()
print(clean_code)  # 예상 출력 / Expected: JPY

# TODO 2: 고객 이름을 Title Case로 변환하세요 / Convert the name to Title Case
raw_name = "park jiwoo"
clean_name = raw_name.title()
print(clean_name)  # 예상 출력 / Expected: Park Jiwoo
```

### 1-2. `strip()` / `lstrip()` / `rstrip()` — 공백 제거

🇰🇷 문자열 양쪽 끝의 공백(또는 지정한 문자)을 제거합니다. 사용자 입력을 받을 때 거의 항상 함께 쓰입니다. 가운데 공백은 제거하지 않는다는 점에 주의하세요!

🇺🇸 Removes whitespace (or specified characters) from the ends of a string. Almost always used when handling user input. Note: spaces **in the middle** are NOT removed!

| 메서드 Method | 설명 Description |
|---|---|
| `strip()` | 양쪽 끝 제거 / both ends |
| `lstrip()` | 왼쪽만 제거 / left only |
| `rstrip()` | 오른쪽만 제거 / right only |


```python
# 데모 / Demo — 사용자가 입력한 송금 메모 정리
memo = "   rent payment   "

print(f"[{memo.strip()}]")    # [rent payment]
print(f"[{memo.lstrip()}]")   # [rent payment   ]
print(f"[{memo.rstrip()}]")   # [   rent payment]

# 특정 문자 제거도 가능 / You can also strip specific characters
amount_text = "$$1500$$"
print(amount_text.strip("$"))  # 1500
```

```python
# 연습 / Practice
# TODO 1: 입력값의 양쪽 공백을 제거하세요 / Strip whitespace from both ends
user_input = "  WONDER PAY  "
clean_input = user_input.strip()
print(f"[{clean_input}]")  # 예상 출력 / Expected: [WONDER PAY]

# TODO 2: 문자열 끝의 줄바꿈 문자(\n)를 제거하세요 / Remove the trailing newline
line = "balance: 50000\n"
clean_line = line.strip("\n")
print(f"[{clean_line}]")  # 예상 출력 / Expected: [balance: 50000]
```

### 1-3. `split()` — 문자열 쪼개기

🇰🇷 문자열을 구분자(separator) 기준으로 잘라 **리스트로 반환**합니다. 구분자를 생략하면 공백 기준으로 자릅니다. CSV 데이터 한 줄을 처리할 때 필수입니다.

🇺🇸 Splits a string by a separator and **returns a list**. If no separator is given, it splits on whitespace. Essential for processing a line of CSV data.


```python
# 데모 / Demo — Wonder Exchange 환율 데이터 한 줄 처리
csv_line = "USD,1350.50,2026-06-12"
fields = csv_line.split(",")
print(fields)        # ['USD', '1350.50', '2026-06-12']
print(fields[0])     # USD

sentence = "learn python every day"
words = sentence.split()
print(words)         # ['learn', 'python', 'every', 'day']
```

```python
# 연습 / Practice
# TODO 1: 콜론(:)을 기준으로 거래 기록을 쪼개세요 / Split the record by colon
record = "wonder_pay:coffee:4500"
parts = record.split(":")
print(parts)  # 예상 출력 / Expected: ['wonder_pay', 'coffee', '4500']

# TODO 2: 문장을 단어 리스트로 쪼개세요 / Split the sentence into words
quote = "practice makes perfect"
word_list = quote.split()
print(len(word_list))  # 예상 출력 / Expected: 3
```

### 1-4. `join()` — 리스트를 문자열로 합치기

🇰🇷 `split()`의 반대입니다. 리스트의 요소들을 구분자로 이어 붙여 **하나의 문자열**을 만듭니다. 주의: 구분자가 앞에 오고, 리스트가 괄호 안에 들어갑니다 — `"구분자".join(리스트)`. 요소는 모두 문자열이어야 합니다!

🇺🇸 The opposite of `split()`. Joins list elements into **one string** with a separator. Watch the syntax: the separator comes first — `"sep".join(my_list)`. All elements must be strings!


```python
# 데모 / Demo — 거래 기록을 CSV 형식으로 저장하기
fields = ["KRW", "1.00", "2026-06-12"]
csv_line = ",".join(fields)
print(csv_line)  # KRW,1.00,2026-06-12

words = ["wonder", "exchange"]
print(" ".join(words))   # wonder exchange
print("-".join(words))   # wonder-exchange
```

```python
# 연습 / Practice
# TODO 1: 단어들을 공백으로 이어 붙이세요 / Join the words with spaces
words = ["final", "exam", "fighting"]
sentence = " ".join(words)
print(sentence)  # 예상 출력 / Expected: final exam fighting

# TODO 2: 계좌번호 조각을 하이픈(-)으로 연결하세요 / Join account parts with hyphens
account_parts = ["110", "456", "789012"]
account_number = "-".join(account_parts)
print(account_number)  # 예상 출력 / Expected: 110-456-789012
```

### 1-5. `replace()` — 문자열 바꾸기

🇰🇷 문자열 안의 특정 부분을 다른 문자열로 바꾼 **새 문자열을 반환**합니다. 모든 일치 항목이 바뀝니다. 숫자에서 쉼표를 제거할 때(`"1,350,000"` → `"1350000"`) 매우 유용합니다.

🇺🇸 Returns a **new string** with all occurrences of a substring replaced. Very useful for removing commas from numbers (`"1,350,000"` → `"1350000"`).


```python
# 데모 / Demo — 환율 표기에서 쉼표 제거하기
price_text = "1,350,000"
clean_text = price_text.replace(",", "")
print(clean_text)        # 1350000
print(int(clean_text))   # 1350000 (이제 숫자로 변환 가능!)

greeting = "hello hello hello"
print(greeting.replace("hello", "bye"))  # bye bye bye
```

### 1-6. `find()` / `index()` / `count()` — 찾기와 세기

🇰🇷 부분 문자열의 위치를 찾거나 등장 횟수를 셉니다. **핵심 차이:** 찾는 문자열이 없을 때 `find()`는 `-1`을 반환하지만, `index()`는 **에러(ValueError)**를 발생시킵니다!

🇺🇸 Find the position of a substring or count its occurrences. **Key difference:** when the substring is missing, `find()` returns `-1`, but `index()` raises a **ValueError**!

| 메서드 Method | 없을 때 When missing |
|---|---|
| `find()` | `-1` 반환 / returns `-1` |
| `index()` | 에러 발생 / raises error |
| `count()` | `0` 반환 / returns `0` |


```python
# 데모 / Demo — 거래 메모에서 키워드 찾기
memo = "coffee at wonder cafe, coffee beans"

print(memo.find("wonder"))    # 0  (첫 번째 위치 / first position)
print(memo.find("tea"))       # -1 (없음 / not found)
print(memo.index("wonder"))   # 10
print(memo.count("coffee"))   # 2

# memo.index("tea")  # 주석을 풀면 ValueError 발생! / Uncomment → ValueError!
```

```python
# 연습 / Practice
# TODO 1: 'pay'가 처음 나타나는 위치를 찾으세요 / Find the first position of 'pay'
service_name = "wonder pay"
position = service_name.find("pay")
print(position)  # 예상 출력 / Expected: 7

# TODO 2: 문자 'o'가 몇 번 나오는지 세세요 / Count how many times 'o' appears
position_count = service_name.count("o")
print(position_count)  # 예상 출력 / Expected: 1
```

### 1-7. `startswith()` / `endswith()` — 시작과 끝 확인

🇰🇷 문자열이 특정 문자열로 시작하는지/끝나는지 확인하여 `True` 또는 `False`를 반환합니다. 파일 확장자 확인(`.csv`)이나 계좌번호 앞자리 확인에 자주 쓰입니다.

🇺🇸 Checks whether a string starts/ends with a given substring, returning `True` or `False`. Common for checking file extensions (`.csv`) or account number prefixes.


```python
# 데모 / Demo — 파일 이름과 계좌번호 확인
file_name = "exchange_rates.csv"
print(file_name.endswith(".csv"))   # True
print(file_name.endswith(".json"))  # False

account_number = "110-456-789012"
print(account_number.startswith("110"))  # True
```

### 1-8. `isdigit()` / `isalpha()` / `isalnum()` — 문자 종류 검사

🇰🇷 문자열이 어떤 종류의 문자로만 이루어졌는지 검사합니다. 사용자가 입력한 금액이 진짜 숫자인지 확인할 때 필수입니다. 빈 문자열은 항상 `False`입니다!

🇺🇸 Tests what kind of characters a string contains. Essential for validating that a user-entered amount is really a number. An empty string always returns `False`!

| 메서드 Method | True 조건 / True when |
|---|---|
| `isdigit()` | 숫자만 / digits only |
| `isalpha()` | 글자만 / letters only |
| `isalnum()` | 숫자+글자만 / digits and letters only |


```python
# 데모 / Demo — 송금 금액 입력값 검증
amount_input = "50000"
print(amount_input.isdigit())   # True

bad_input = "5만원"
print(bad_input.isdigit())      # False ('만원'은 숫자가 아님)

user_id = "minji2026"
print(user_id.isalnum())        # True
print(user_id.isalpha())        # False (숫자 포함 / contains digits)
```

```python
# 연습 / Practice
# TODO 1: 입력값이 숫자로만 이루어졌는지 확인하세요 / Check if the input is all digits
pin_input = "1234"
is_valid_pin = pin_input.isdigit()
print(is_valid_pin)  # 예상 출력 / Expected: True

# TODO 2: 이름이 글자로만 이루어졌는지 확인하세요 / Check if the name is letters only
name_input = "minji"
is_valid_name = name_input.isalpha()
print(is_valid_name)  # 예상 출력 / Expected: True
```

### 1-9. `zfill()` / f-string 복습 — 자릿수 채우기와 포매팅

🇰🇷 `zfill(n)`은 문자열 왼쪽을 0으로 채워 전체 길이를 n으로 만듭니다. 주문번호나 회원번호 형식을 맞출 때 유용합니다. f-string은 메서드는 아니지만 시험에 꼭 나오니 함께 복습하세요!

🇺🇸 `zfill(n)` pads the left side with zeros until the string is n characters long. Useful for formatting order numbers. f-strings aren't methods, but they're guaranteed exam material — review them together!


```python
# 데모 / Demo — Wonder Pay 주문번호 만들기
order_id = "42"
print(order_id.zfill(6))  # 000042

# f-string 복습 / f-string review
balance = 1234567.891
print(f"잔액: {balance:,.2f}원")   # 잔액: 1,234,567.89원
print(f"{'KRW':>5}")               # "  KRW" (오른쪽 정렬 / right-aligned)
```

## 2️⃣ 리스트 메서드 / List Methods

리스트는 **수정 가능(mutable)**합니다. 대부분의 리스트 메서드는 원본을 **직접 수정**하고 `None`을 반환합니다 — 문자열과 정반대입니다! 🔥

Lists are **mutable**. Most list methods **modify the original in place** and return `None` — the opposite of strings! 🔥

### 2-1. `append()` / `insert()` — 요소 추가하기

🇰🇷 `append(x)`는 리스트 **맨 뒤에** 요소 하나를 추가하고, `insert(i, x)`는 **원하는 위치 i에** 끼워 넣습니다. `append`는 리스트를 직접 수정하며 `None`을 반환하니, `my_list = my_list.append(x)`처럼 쓰면 안 됩니다!

🇺🇸 `append(x)` adds one element **to the end**; `insert(i, x)` inserts it **at position i**. `append` modifies the list in place and returns `None` — never write `my_list = my_list.append(x)`!


```python
# 데모 / Demo — 오늘의 환율 기록 쌓기
usd_rates = [1349.2, 1350.5]
usd_rates.append(1351.0)
print(usd_rates)  # [1349.2, 1350.5, 1351.0]

usd_rates.insert(0, 1348.8)  # 맨 앞에 삽입 / insert at the front
print(usd_rates)  # [1348.8, 1349.2, 1350.5, 1351.0]
```

```python
# 연습 / Practice
# TODO 1: 거래 내역에 'taxi'를 맨 뒤에 추가하세요 / Append 'taxi' to the end
purchases = ["coffee", "lunch"]
purchases.append("taxi")
print(purchases)  # 예상 출력 / Expected: ['coffee', 'lunch', 'taxi']

# TODO 2: 인덱스 1 위치에 'snack'을 끼워 넣으세요 / Insert 'snack' at index 1
purchases.insert(1, "snack")
print(purchases)  # 예상 출력 / Expected: ['coffee', 'snack', 'lunch', 'taxi']
```

### 2-2. `extend()` — 리스트 이어 붙이기

🇰🇷 다른 리스트의 **모든 요소를** 뒤에 이어 붙입니다. `append(리스트)`와 헷갈리지 마세요 — `append`는 리스트 전체를 **요소 하나로** 넣어버립니다! 이건 시험 단골 함정입니다. 🔥

🇺🇸 Adds **all elements** of another list to the end. Don't confuse it with `append(a_list)` — `append` would add the whole list as **one single element**! Classic exam trap. 🔥


```python
# 데모 / Demo — extend vs append 차이 (시험 포인트!)
morning_rates = [1349.2, 1350.5]
afternoon_rates = [1351.0, 1352.3]

list_a = morning_rates.copy()
list_a.extend(afternoon_rates)
print(list_a)  # [1349.2, 1350.5, 1351.0, 1352.3]  ← 요소 4개

list_b = morning_rates.copy()
list_b.append(afternoon_rates)
print(list_b)  # [1349.2, 1350.5, [1351.0, 1352.3]]  ← 요소 3개 (리스트가 통째로!)
```

```python
# 연습 / Practice
# TODO 1: 두 번째 리스트의 모든 요소를 이어 붙이세요 / Extend with all elements
week1_spending = [4500, 12000]
week2_spending = [8000, 3500]
week1_spending.extend(week2_spending)
print(week1_spending)       # 예상 출력 / Expected: [4500, 12000, 8000, 3500]
print(len(week1_spending))  # 예상 출력 / Expected: 4
```

### 2-3. `remove()` / `pop()` — 요소 삭제하기

🇰🇷 `remove(x)`는 **값 x를 찾아서** 첫 번째 것만 삭제합니다(반환값 없음). `pop(i)`는 **인덱스 i의 요소를** 삭제하고 **그 값을 반환**합니다. 인덱스를 생략하면 마지막 요소를 꺼냅니다.

🇺🇸 `remove(x)` deletes the **first occurrence of the value** x (returns nothing). `pop(i)` deletes the element **at index i** and **returns it**. Without an index, it pops the last element.

| 메서드 Method | 기준 By | 반환 Returns |
|---|---|---|
| `remove(x)` | 값 / value | `None` |
| `pop(i)` | 인덱스 / index | 삭제된 값 / removed value |


```python
# 데모 / Demo — 장바구니에서 항목 빼기
cart = ["coffee", "cake", "coffee", "juice"]

cart.remove("coffee")   # 첫 번째 'coffee'만 삭제
print(cart)             # ['cake', 'coffee', 'juice']

last_item = cart.pop()  # 마지막 요소를 꺼내서 반환
print(last_item)        # juice
print(cart)             # ['cake', 'coffee']

first_item = cart.pop(0)
print(first_item)       # cake
```

### 2-5. `sort()` / `reverse()` vs `sorted()` — 정렬하기

🇰🇷 `sort()`는 리스트를 **제자리에서(in-place)** 정렬하고 `None`을 반환합니다. 내장 함수 `sorted()`는 원본은 그대로 두고 **정렬된 새 리스트를 반환**합니다. `reverse=True`로 내림차순 정렬이 가능합니다. `reverse()`는 정렬이 아니라 단순히 순서를 뒤집습니다!

🇺🇸 `sort()` sorts the list **in place** and returns `None`. The built-in `sorted()` leaves the original alone and **returns a new sorted list**. Use `reverse=True` for descending order. `reverse()` just flips the order — it does NOT sort!

| 구분 | `sort()` | `sorted()` |
|---|---|---|
| 원본 변경 / Changes original | ✅ | ❌ |
| 반환값 / Returns | `None` | 새 리스트 / new list |


```python
# 데모 / Demo — 일주일 지출 정렬하기
spending = [12000, 4500, 30000, 8000]

sorted_copy = sorted(spending, reverse=True)
print(sorted_copy)  # [4500, 8000, 12000, 30000]
print(spending)     # [12000, 4500, 30000, 8000]  ← 원본 그대로!

spending.sort(reverse=True)  # 내림차순, 제자리 정렬
print(spending)     # [30000, 12000, 8000, 4500]

names = ["c", "a", "b"]
names.reverse()     # 정렬 아님! 그냥 뒤집기
print(names)        # ['b', 'a', 'c']
```

### 2-6. `copy()` / `clear()` — 복사와 비우기

🇰🇷 `copy()`는 리스트의 **복사본**을 만듭니다. `list_b = list_a`라고 쓰면 복사가 아니라 **같은 리스트에 이름 두 개**를 붙이는 것이니 주의하세요! (시험 단골 함정 🔥) `clear()`는 모든 요소를 삭제해 빈 리스트로 만듭니다.

🇺🇸 `copy()` makes a **copy** of the list. Writing `list_b = list_a` does NOT copy — it gives **two names to the same list**! (Classic exam trap 🔥) `clear()` removes everything, leaving an empty list.


```python
# 데모 / Demo — 복사 함정 (시험 포인트!)
original = [100, 200]

alias = original          # 복사 아님! 같은 리스트
alias.append(300)
print(original)           # [100, 200, 300]  ← 원본도 바뀜!

real_copy = original.copy()
real_copy.append(999)
print(original)           # [100, 200, 300]  ← 원본은 안전
print(real_copy)          # [100, 200, 300, 999]

cart = ["coffee", "cake"]
cart.clear()
print(cart)               # []
```

## 3️⃣ 딕셔너리 메서드 / Dictionary Methods

딕셔너리는 **키(key) → 값(value)** 쌍을 저장합니다. 키는 중복될 수 없습니다.

Dictionaries store **key → value** pairs. Keys cannot be duplicated.

### 3-1. `keys()` / `values()` / `items()` — 딕셔너리 순회 3총사

🇰🇷 딕셔너리를 for문으로 돌 때 쓰는 세 가지 메서드입니다. `items()`는 (키, 값) 쌍을 주기 때문에 변수 두 개로 받을 수 있습니다 — 시험에 가장 자주 나오는 패턴입니다!

🇺🇸 The three methods for looping over a dictionary. `items()` gives (key, value) pairs, so you can unpack into two variables — the most common exam pattern!

| 메서드 Method | 반환 Returns |
|---|---|
| `keys()` | 키들 / all keys |
| `values()` | 값들 / all values |
| `items()` | (키, 값) 쌍들 / (key, value) pairs |


```python
# 데모 / Demo — Wonder Exchange 오늘의 환율표
exchange_rates = {"USD": 1350.5, "EUR": 1462.3, "JPY": 8.65}

print(list(exchange_rates.keys()))    # ['USD', 'EUR', 'JPY']
print(list(exchange_rates.values()))  # [1350.5, 1462.3, 8.65]

for currency, rate in exchange_rates.items():
    print(f"{currency}: {rate}")
# USD: 1350.5
# EUR: 1462.3
# JPY: 8.65
```

```python
# 연습 / Practice
menu_prices = {"americano": 4500, "latte": 5000, "cake": 6500}

# TODO 1: 모든 메뉴 이름(키)을 리스트로 출력하세요 / Print all menu names as a list
print(list(menu_prices.keys()))  # 예상 출력 / Expected: ['americano', 'latte', 'cake']

# TODO 2: items()로 순회하며 "메뉴: 가격" 형태로 출력하세요 / Loop with items()
for menu_name, price in menu_prices.items():
    print(f"{menu_name}: {price}")
# 예상 출력 / Expected:
# americano: 4500
# latte: 5000
# cake: 6500
```

### 3-2. `get()` — 안전하게 값 꺼내기

🇰🇷 `dict[키]`는 키가 없으면 **에러(KeyError)**가 나지만, `get(키)`는 `None`을 반환합니다. `get(키, 기본값)`으로 기본값도 지정할 수 있습니다. 횟수 세기 패턴 `count_dict[key] = count_dict.get(key, 0) + 1`은 반드시 외워두세요! 🔥

🇺🇸 `dict[key]` raises a **KeyError** if the key is missing, but `get(key)` returns `None`. You can also give a default: `get(key, default)`. Memorize the counting pattern `count_dict[key] = count_dict.get(key, 0) + 1`! 🔥


```python
# 데모 / Demo — 잔액 조회와 횟수 세기 패턴
balances = {"minji": 50000, "jiwoo": 32000}

print(balances.get("minji"))        # 50000
print(balances.get("haram"))        # None (에러 없음!)
print(balances.get("haram", 0))     # 0 (기본값 / default)

# 횟수 세기 패턴 / The counting pattern
purchases = ["coffee", "cake", "coffee"]
count_dict = {}
for item in purchases:
    count_dict[item] = count_dict.get(item, 0) + 1
print(count_dict)  # {'coffee': 2, 'cake': 1}
```

```python
# 연습 / Practice
stock = {"apple": 10, "banana": 5}

# TODO 1: 'cherry'의 재고를 조회하되, 없으면 0을 반환하세요 / Get 'cherry' stock, default 0
cherry_stock = stock.get("cherry", 0)
print(cherry_stock)  # 예상 출력 / Expected: 0

# TODO 2: get()을 이용해 통화별 거래 횟수를 세세요 / Count transactions per currency
transactions = ["USD", "KRW", "USD", "USD"]
currency_count = {}
for currency in transactions:
    currency_count[currency] = currency_count.get(currency, 0) + 1
print(currency_count)  # 예상 출력 / Expected: {'USD': 3, 'KRW': 1}
```

### 3-3. `update()` / `pop()` — 합치기와 삭제하기

🇰🇷 `update(다른_딕셔너리)`는 여러 키-값 쌍을 한 번에 추가/수정합니다. **이미 있는 키는 새 값으로 덮어씁니다!** `pop(키)`는 해당 키를 삭제하고 **값을 반환**합니다. `pop(키, 기본값)`으로 키가 없을 때의 에러를 막을 수 있습니다.

🇺🇸 `update(other_dict)` adds/updates several key-value pairs at once. **Existing keys get overwritten!** `pop(key)` deletes the key and **returns its value**. Use `pop(key, default)` to avoid an error when the key is missing.


```python
# 데모 / Demo — 환율표 갱신과 항목 삭제
rates = {"USD": 1350.5, "EUR": 1462.3}

rates.update({"JPY": 8.65, "USD": 1352.0})  # JPY 추가, USD 덮어쓰기
print(rates)  # {'USD': 1352.0, 'EUR': 1462.3, 'JPY': 8.65}

removed_rate = rates.pop("EUR")
print(removed_rate)  # 1462.3
print(rates)         # {'USD': 1352.0, 'JPY': 8.65}

print(rates.pop("GBP", "없음"))  # 없음 (에러 대신 기본값)
```

### 3-4. `in` 연산자와 `setdefault()` — 키 확인과 기본값 설정

🇰🇷 `키 in 딕셔너리`는 **키의 존재 여부만** 확인합니다 (값이 아님!). `setdefault(키, 기본값)`은 키가 없으면 기본값으로 추가하고, 있으면 기존 값을 그대로 반환합니다.

🇺🇸 `key in my_dict` checks **only keys**, not values! `setdefault(key, default)` adds the key with the default if missing, or returns the existing value if present.


```python
# 데모 / Demo — 회원 확인하기
members = {"minji": "gold", "jiwoo": "silver"}

print("minji" in members)   # True
print("gold" in members)    # False (값은 검사 안 함! / values not checked!)

members.setdefault("haram", "bronze")  # 없으니 추가됨
members.setdefault("minji", "bronze")  # 이미 있으니 그대로
print(members)  # {'minji': 'gold', 'jiwoo': 'silver', 'haram': 'bronze'}
```

## 4️⃣ 튜플 / Tuples

튜플은 **수정 불가능한** 순서 자료형입니다. 메서드는 단 두 개 — 대신 언패킹이 핵심입니다!

Tuples are **immutable** sequences. Only two methods — unpacking is the real star!

### 4-1. 튜플의 특징과 `count()` / `index()`

🇰🇷 튜플은 **수정 불가능(immutable)**한 리스트입니다. `append`, `remove` 같은 수정 메서드가 없고, 오직 `count()`와 `index()` 두 개만 있습니다. 바뀌면 안 되는 데이터(좌표, 설정값, 환율 기준일)에 사용합니다.

🇺🇸 A tuple is an **immutable** list. It has no modifying methods like `append` or `remove` — only `count()` and `index()`. Use it for data that must not change (coordinates, settings, rate dates).


```python
# 데모 / Demo — 수정 불가능한 거래 기록
transaction = ("USD", 1350.5, "2026-06-12", "USD")

print(transaction.count("USD"))   # 2
print(transaction.index(1350.5))  # 1

# transaction[0] = "EUR"     # 주석을 풀면 TypeError! / Uncomment → TypeError!
# transaction.append("new")  # AttributeError! 튜플에는 append가 없음
```

### 4-2. 튜플 언패킹(Unpacking) — 한 번에 여러 변수에 담기

🇰🇷 튜플의 요소들을 **여러 변수에 한 번에** 나눠 담을 수 있습니다. 함수가 여러 값을 반환할 때(사실은 튜플 하나!) 늘 쓰는 문법입니다. 변수 개수와 요소 개수가 맞아야 합니다.

🇺🇸 You can unpack tuple elements **into several variables at once**. This is exactly what happens when a function returns multiple values (it's really one tuple!). The number of variables must match the number of elements.


```python
# 데모 / Demo — 환율 정보 언패킹
rate_info = ("USD", 1350.5)
currency, rate = rate_info
print(currency)  # USD
print(rate)      # 1350.5

# 함수의 다중 반환값도 사실 튜플! / Multiple return values are really a tuple!
def get_min_max(numbers):
    return min(numbers), max(numbers)

low_rate, high_rate = get_min_max([1350, 1348, 1355])
print(low_rate, high_rate)  # 1348 1355

# 변수 교환도 한 줄로 / Swap in one line
a, b = 10, 20
a, b = b, a
print(a, b)  # 20 10
```

## 5️⃣ 집합 메서드 / Set Methods

집합은 **중복 없음 + 순서 없음**. 인덱싱(`s[0]`)이 불가능하다는 점에 주의하세요!

Sets: **no duplicates + no order**. Note that indexing (`s[0]`) is impossible!

### 5-1. `add()` / `update()` — 집합에 추가하기

🇰🇷 집합(set)은 **중복이 없고 순서가 없는** 자료형입니다. `add(x)`는 요소 하나를, `update(iterable)`은 여러 요소를 추가합니다. 이미 있는 값을 추가하면 조용히 무시됩니다 — 에러가 나지 않아요!

🇺🇸 A set has **no duplicates and no order**. `add(x)` adds one element; `update(iterable)` adds several. Adding an existing value is silently ignored — no error!


```python
# 데모 / Demo — 오늘 거래된 통화 모으기
traded_currencies = {"USD", "KRW"}

traded_currencies.add("EUR")
traded_currencies.add("USD")  # 이미 있음 → 무시됨
print(sorted(traded_currencies))  # ['EUR', 'KRW', 'USD']

traded_currencies.update(["JPY", "CNY"])
print(len(traded_currencies))  # 5
```

### 5-2. `remove()` / `discard()` — 집합에서 삭제하기

🇰🇷 둘 다 요소를 삭제하지만, **없는 값을 삭제할 때** 행동이 다릅니다: `remove()`는 **에러(KeyError)**, `discard()`는 **조용히 넘어감**. 이 차이는 시험 단골입니다! 🔥

🇺🇸 Both delete an element, but they differ **when the value is missing**: `remove()` raises a **KeyError**, while `discard()` does **nothing quietly**. A classic exam point! 🔥


```python
# 데모 / Demo — remove vs discard
active_users = {"minji", "jiwoo", "haram"}

active_users.remove("haram")
print(sorted(active_users))     # ['jiwoo', 'minji']

active_users.discard("ghost")   # 없지만 에러 없음!
print(sorted(active_users))     # ['jiwoo', 'minji']

# active_users.remove("ghost")  # 주석을 풀면 KeyError!
```

### 5-3. 집합 연산 — `union()` / `intersection()` / `difference()`

🇰🇷 두 집합을 비교하는 강력한 연산들입니다. 메서드 대신 연산자 기호(`|`, `&`, `-`)도 쓸 수 있습니다. "두 지점에서 모두 거래된 통화는?" 같은 문제를 한 줄로 해결합니다!

🇺🇸 Powerful operations for comparing two sets. You can use operator symbols (`|`, `&`, `-`) instead of method names. Questions like "which currencies were traded at BOTH branches?" become one-liners!

| 메서드 Method | 기호 Symbol | 의미 Meaning |
|---|---|---|
| `union()` | `\|` | 합집합 (둘 다 합침) / all from both |
| `intersection()` | `&` | 교집합 (공통) / common to both |
| `difference()` | `-` | 차집합 (앞에만 있음) / in first only |
| `symmetric_difference()` | `^` | 한쪽에만 있음 / in exactly one |


```python
# 데모 / Demo — 두 지점의 거래 통화 비교
seoul_branch = {"USD", "EUR", "JPY"}
busan_branch = {"USD", "CNY"}

print(sorted(seoul_branch.union(busan_branch)))         # ['CNY', 'EUR', 'JPY', 'USD']
print(sorted(seoul_branch.intersection(busan_branch)))  # ['USD']
print(sorted(seoul_branch.difference(busan_branch)))    # ['EUR', 'JPY']
print(sorted(seoul_branch ^ busan_branch))              # ['CNY', 'EUR', 'JPY']

# 기호 버전도 동일 / Symbols work the same
print(sorted(seoul_branch & busan_branch))              # ['USD']
```

```python
# 연습 / Practice
python_class = {"minji", "jiwoo", "haram"}
math_class = {"jiwoo", "sora"}

# TODO 1: 두 수업을 모두 듣는 학생을 찾으세요 / Students taking BOTH classes
both_classes = python_class.intersection(math_class)
print(sorted(both_classes))  # 예상 출력 / Expected: ['jiwoo']

# TODO 2: 파이썬 수업만 듣는 학생을 찾으세요 / Students taking ONLY Python
only_python = python_class.difference(math_class)
print(sorted(only_python))  # 예상 출력 / Expected: ['haram', 'minji']

# TODO 3: 전체 학생 명단을 만드세요 / All students combined
all_students = python_class.union(math_class)
print(sorted(all_students))  # 예상 출력 / Expected: ['haram', 'jiwoo', 'minji', 'sora']
```

### 5-4. `issubset()` / `issuperset()` + 중복 제거 활용

🇰🇷 `issubset()`은 "내가 상대방 안에 모두 포함되는가?", `issuperset()`은 "내가 상대방을 모두 포함하는가?"를 확인합니다. 그리고 `set()`으로 변환하면 리스트의 중복을 가장 쉽게 제거할 수 있습니다 — 이번 프로젝트의 `remove_duplicates` 문제와 연결되는 개념입니다!

🇺🇸 `issubset()` asks "am I fully contained in the other set?"; `issuperset()` asks "do I contain all of the other set?". Also, converting to `set()` is the easiest way to remove duplicates from a list — connecting back to our `remove_duplicates` problem!


```python
# 데모 / Demo — 권한 확인과 중복 제거
required_permissions = {"read", "transfer"}
user_permissions = {"read", "transfer", "admin"}

print(required_permissions.issubset(user_permissions))    # True
print(user_permissions.issuperset(required_permissions))  # True

# 리스트 중복 제거 한 줄 패턴 / One-line dedup pattern
visit_log = ["minji", "jiwoo", "minji", "minji"]
unique_visitors = list(set(visit_log))
print(len(unique_visitors))  # 2
```

```python
# 연습 / Practice
basic_features = {"login", "transfer"}
premium_features = {"login", "transfer", "invest", "loan"}

# TODO 1: 기본 기능이 프리미엄에 모두 포함되는지 확인하세요
#         / Check if basic features are all included in premium
is_included = basic_features.issubset(premium_features)
print(is_included)  # 예상 출력 / Expected: True

# TODO 2: set을 이용해 중복 없는 통화 개수를 구하세요 / Count unique currencies using set
trade_log = ["USD", "KRW", "USD", "JPY", "KRW"]
unique_count = len(set(trade_log))
print(unique_count)  # 예상 출력 / Expected: 3
```
