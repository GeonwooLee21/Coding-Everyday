# 📦 1. 리스트 (List) — 🛒 편의점 장바구니

편의점에서 장을 보고 있습니다. 장바구니에 담긴 상품과 가격을 리스트로 관리해 봅시다.
You're shopping at a convenience store. Let's manage the cart items and prices with lists.

## 📋 핵심 연산 요약 / Core Operations

| 연산 / Operation | 문법 / Syntax | 설명 / Description |
|---|---|---|
| 생성 / Create | `items = ["a", "b"]` | 대괄호 사용 / square brackets |
| 인덱싱 / Indexing | `items[0]`, `items[-1]` | 첫 번째 / 마지막 요소 |
| 슬라이싱 / Slicing | `items[1:3]` | 1번부터 2번까지 (3번 제외) |
| 끝에 추가 / Append | `items.append(x)` | 맨 뒤에 추가 |
| 중간에 삽입 / Insert | `items.insert(i, x)` | i번 위치에 삽입 |
| 값으로 삭제 / Remove | `items.remove(x)` | 처음 만나는 x 삭제 |
| 정렬 / Sort | `items.sort()` | 원본을 직접 정렬 (반환값 `None`!) |
| 집계 / Aggregate | `sum()`, `len()`, `max()`, `min()` | 합계, 길이, 최대, 최소 |
| 위치 찾기 / Find | `items.index(x)` | x가 처음 나오는 인덱스 |
| 포함 확인 / Membership | `x in items` | `True` / `False` |

## 👀 따라해 보기 / Worked Examples

아래 두 셀을 실행하면서 각 줄이 무엇을 하는지 확인하세요.
Run the two cells below and trace what each line does.

````python
# 예제 1: 장바구니에 담고, 빼고 / Example 1: add and remove items
cart_items = ["삼각김밥", "바나나우유"]
cart_items.append("컵라면")        # 맨 뒤에 추가 / add to the end
cart_items.insert(0, "도시락")      # 맨 앞에 삽입 / insert at the front
print(cart_items)                   # ['도시락', '삼각김밥', '바나나우유', '컵라면']

cart_items.remove("삼각김밥")       # 값으로 삭제 / remove by value
print(cart_items)                   # ['도시락', '바나나우유', '컵라면']
print(len(cart_items))              # 3
````

````python
# 예제 2: 가격 집계 / Example 2: price aggregation
cart_prices = [4500, 1800, 1100]
print(sum(cart_prices))                      # 7400 — 총액 / total
print(max(cart_prices))                      # 4500 — 최고가 / highest
print(sum(cart_prices) / len(cart_prices))   # 2466.666... — 평균 / average
print(cart_prices[:2])                       # [4500, 1800] — 앞 2개 / first two
````

````python
# ─────────────────────────────────────────
# [L5] 비싼 순으로 정렬 / Sort by price, descending
# 가격을 비싼 순서(내림차순)로 정렬하세요. (원본을 직접 수정!)
# Sort prices from highest to lowest, IN-PLACE.
# ─────────────────────────────────────────
cart_prices = [1200, 1800, 1100, 1500]
cart_prices.sort(reverse=True)
print(cart_prices)

# 예상 출력 / Expected output:
# [1800, 1500, 1200, 1100]
````

````python
# ─────────────────────────────────────────
# [L6] 총액과 평균 / Total and average
# 장바구니 총액과 평균 가격을 계산하세요.
# Calculate the total and the average price.
# ─────────────────────────────────────────
cart_prices = [1200, 1800, 1100, 1500]
total_price = sum(cart_prices)
average_price = total_price / len(cart_prices)
print(f"총액: {total_price}원, 평균: {average_price}원")

# 예상 출력 / Expected output:
# 총액: 5600원, 평균: 1400.0원
````

````python
# ─────────────────────────────────────────
# [L7] 비싼 상품 세기 / Count expensive items
# 반복문으로 1500원 이상인 상품의 개수를 세어보세요.
# Use a loop to count items priced 1500 won OR MORE.
# (경계 조건 주의! >= 인가 > 인가? / Watch the boundary: >= or >?)
# ─────────────────────────────────────────
cart_prices = [1200, 1800, 1100, 1500]
expensive_count = 0
for price in cart_prices:
    if price >= 1500:
        expensive_count += 1
print(f"1500원 이상 상품: {expensive_count}개")

# 예상 출력 / Expected output:
# 1500원 이상 상품: 2개
````

````python
# ─────────────────────────────────────────
# [L8] 최고가 상품의 위치 / Where is the most expensive item?
# 최고가를 찾고, 그 가격이 리스트의 몇 번 인덱스에 있는지 찾으세요.
# Find the highest price and its index in the list.
# ─────────────────────────────────────────
cart_prices = [1200, 1800, 1100, 1500]
max_price = max(cart_prices)
max_index = cart_prices.index(max_price)
print(f"최고가 {max_price}원은 인덱스 {max_index}에 있습니다")

# 예상 출력 / Expected output:
# 최고가 1800원은 인덱스 1에 있습니다
````

## ⚠️ 함정 주의 / Watch Out! — 리스트 편

시험에 정말 자주 나오는 함정 두 가지입니다. 아래 셀을 실행하고 출력이 왜 그런지 설명할 수 있어야 합니다.
Two traps that show up on exams all the time. Run the cell and make sure you can explain WHY.

````python
# 함정 1: 별명(alias) vs 복사(copy)
# Trap 1: alias vs copy
cart_a = [1000, 2000]
cart_b = cart_a            # ⚠️ 복사가 아니라 같은 리스트의 '별명'입니다!
cart_b.append(3000)        #    NOT a copy — just another name for the SAME list!
print("cart_a:", cart_a)   # cart_a도 바뀌었어요?! / cart_a changed too?!

cart_c = cart_a.copy()     # ✅ 진짜 복사 / a real copy (cart_a[:] also works)
cart_c.append(4000)
print("cart_a:", cart_a)   # 이번엔 안 바뀜 / unchanged this time
print("cart_c:", cart_c)

# 함정 2: sort()의 반환값은 None!
# Trap 2: sort() returns None!
result = cart_a.sort()
print("result:", result)   # None! sort()는 원본을 바꾸고 아무것도 반환하지 않습니다
print("cart_a:", cart_a)   # 정렬은 원본에 적용됨 / the sorting happened in-place
````

# 📌 2. 튜플 (Tuple) — 💱 Wonder Exchange 환율 스냅샷

Wonder Exchange에서는 특정 시점의 환율을 `(통화코드, 환율)` 형태로 기록합니다. 한번 기록된 스냅샷은 **변경되면 안 되기 때문에** 튜플이 딱 맞는 자료형입니다!
Wonder Exchange records exchange rates as `(currency_code, rate)` snapshots. Once recorded, a snapshot **must never change** — that's exactly what tuples are for!

## 📋 핵심 연산 요약 / Core Operations

| 연산 / Operation | 문법 / Syntax | 설명 / Description |
|---|---|---|
| 생성 / Create | `t = ("USD", 1352.0)` | 소괄호 사용 / parentheses |
| 인덱싱 / Indexing | `t[0]`, `t[-1]` | 리스트와 동일 / same as lists |
| 언패킹 / Unpacking | `code, rate = t` | 변수에 한 번에 분배 |
| 변경 / Modify | ❌ 불가능! | `TypeError` 발생 / raises `TypeError` |
| 개수 세기 / Count | `t.count(x)` | x가 몇 번 나오는지 |
| 위치 찾기 / Find | `t.index(x)` | x가 처음 나오는 인덱스 |
| 요소 1개 튜플 / Single-element | `(x,)` | **쉼표 필수!** / comma REQUIRED! |
| 변환 / Convert | `list(t)`, `tuple(lst)` | 리스트 ↔ 튜플 |

## 👀 따라해 보기 / Worked Examples

````python
# 예제: 환율 스냅샷 기록과 언패킹 / Example: snapshot and unpacking
usd_snapshot = ("USD", 1352.0)

# 방법 1: 인덱싱 / Method 1: indexing
print(usd_snapshot[0], usd_snapshot[1])    # USD 1352.0

# 방법 2: 언패킹 (더 깔끔!) / Method 2: unpacking (cleaner!)
currency_code, exchange_rate = usd_snapshot
print(f"{currency_code} 환율: {exchange_rate}원")   # USD 환율: 1352.0원

# 튜플 리스트를 순회하며 언패킹 / Loop + unpack a list of tuples
rate_snapshots = [("USD", 1352.0), ("JPY", 9.1)]
for code, rate in rate_snapshots:
    print(f"{code}: {rate}")
````

## ✏️ 연습 문제 / Practice Problems (T1 ~ T8)

````python
# ─────────────────────────────────────────
# [T1] 스냅샷 만들기 / Create a snapshot
# ("USD", 1352.0) 튜플을 만들고, 인덱싱으로 통화코드만 출력하세요.
# Create the tuple ("USD", 1352.0) and print ONLY the currency code via indexing.
# ─────────────────────────────────────────
usd_snapshot = ("USD", 1352.0)
print(usd_snapshot[0])

# 예상 출력 / Expected output:
# USD
````

````python
# ─────────────────────────────────────────
# [T2] 언패킹 / Unpacking
# 튜플을 두 변수로 한 번에 언패킹하세요.
# Unpack the tuple into two variables in ONE line.
# ─────────────────────────────────────────
usd_snapshot = ("USD", 1352.0)
currency_code, exchange_rate = usd_snapshot
print(f"{currency_code}의 환율은 {exchange_rate}원입니다")

# 예상 출력 / Expected output:
# USD의 환율은 1352.0원입니다
````

````python
# ─────────────────────────────────────────
# [T3] 변수 교환 / Swap variables
# 튜플 언패킹으로 두 변수의 값을 한 줄에 교환하세요. (임시 변수 금지!)
# Swap the two variables in ONE line using tuple unpacking. (No temp variable!)
# ─────────────────────────────────────────
krw_amount = 1352000
usd_amount = 1000
krw_amount, usd_amount = usd_amount, krw_amount
print(f"krw_amount: {krw_amount}, usd_amount: {usd_amount}")

# 예상 출력 / Expected output:
# krw_amount: 1000, usd_amount: 1352000
````

````python
# ─────────────────────────────────────────
# [T4] 환율 게시판 출력 / Print the rate board
# 반복문에서 바로 언패킹하여 각 환율을 출력하세요.
# Loop over the snapshots, unpacking DIRECTLY in the for statement.
# ─────────────────────────────────────────
rate_snapshots = [("USD", 1352.0), ("JPY", 9.1), ("EUR", 1480.5)]
for code, rate in rate_snapshots:
    print(f"{code}: {rate}")

# 예상 출력 / Expected output:
# USD: 1352.0
# JPY: 9.1
# EUR: 1480.5
````

````python
# ─────────────────────────────────────────
# [T5] count와 index / count and index
# 1350.0이 몇 번 기록되었는지, 1352.0이 처음 나오는 위치는 어디인지 찾으세요.
# How many times was 1350.0 recorded? At what index does 1352.0 first appear?
# ─────────────────────────────────────────
recorded_rates = (1350.0, 1351.5, 1350.0, 1352.0)
print(recorded_rates.count(1350.0))
print(recorded_rates.index(1352.0))

# 예상 출력 / Expected output:
# 2
# 3
````

````python
# ─────────────────────────────────────────
# [T6] 잘못된 스냅샷 수정 / Fix a wrong snapshot
# 튜플은 직접 수정할 수 없습니다! 리스트로 변환 → 수정 → 다시 튜플로 변환하세요.
# Tuples can't be modified directly! Convert to list → fix → convert back to tuple.
# ─────────────────────────────────────────
wrong_snapshot = ("USD", 1350.0)     # 실제 환율은 1352.5였습니다 / actual rate was 1352.5
temp_list = list(wrong_snapshot)
temp_list[1] = 1352.5
fixed_snapshot = tuple(temp_list)
print(fixed_snapshot)

# 예상 출력 / Expected output:
# ('USD', 1352.5)
````

````python
# ─────────────────────────────────────────
# [T7] 여러 값을 반환하는 함수 / Function returning multiple values
# 함수가 여러 값을 반환하면 사실 '튜플'을 반환하는 것입니다!
# 반환값을 두 변수로 언패킹하세요.
# A function returning multiple values actually returns a TUPLE!
# Unpack the return value into two variables.
# ─────────────────────────────────────────
def get_usd_rate():
    return "USD", 1352.0

code, rate = get_usd_rate()
print(f"{code}: {rate}")

# 예상 출력 / Expected output:
# USD: 1352.0
````

````python
# ─────────────────────────────────────────
# [T8] 중첩 튜플 / Nested tuples
# 환율 게시판에서 JPY의 '환율 숫자'만 인덱싱으로 꺼내세요. (인덱스 두 번!)
# Extract ONLY the JPY rate number using indexing. (Index twice!)
# ─────────────────────────────────────────
rate_board = (("USD", 1352.0), ("JPY", 9.1), ("EUR", 1480.5))
rate_list = [["USD", 1352.0], ["JPY", 9.1], ["EUR", 1480.5]]
jpy_rate = rate_board[1][1]
eur_rate = rate_list[2][1]
print(jpy_rate)
print(eur_rate)

# 예상 출력 / Expected output:
# 9.1
````

## ⚠️ 함정 주의 / Watch Out! — 튜플 편

# 🗂️ 3. 딕셔너리 (Dict) — 💳 Wonder Pay 지갑

Wonder Pay에서는 사용자 이름으로 잔액을 바로 찾을 수 있어야 합니다. `이름 → 잔액` 매핑, 바로 딕셔너리가 필요한 순간입니다!
Wonder Pay needs to look up a user's balance by name instantly. A `name → balance` mapping — that's exactly what dicts are for!

## 📋 핵심 연산 요약 / Core Operations

| 연산 / Operation | 문법 / Syntax | 설명 / Description |
|---|---|---|
| 생성 / Create | `d = {"minji": 50000}` | 중괄호 + `키: 값` |
| 조회 / Access | `d["minji"]` | 키가 없으면 `KeyError`! |
| 안전 조회 / Safe access | `d.get("minji", 0)` | 키가 없으면 기본값 반환 |
| 추가·수정 / Add·Update | `d["junho"] = 32000` | 키가 없으면 추가, 있으면 수정 |
| 삭제 / Delete | `d.pop("minji")` | 삭제하고 값을 반환 |
| 키 목록 / Keys | `d.keys()` | 모든 키 |
| 값 목록 / Values | `d.values()` | 모든 값 |
| 쌍 목록 / Items | `d.items()` | `(키, 값)` 쌍 — 반복문에서 언패킹! |
| 포함 확인 / Membership | `"minji" in d` | **키**를 검사 (값 아님!) |

## 👀 따라해 보기 / Worked Examples

````python
# 예제: 지갑 만들고 조회하기 / Example: build and query the wallet
wallet_balances = {"minji": 50000, "junho": 32000}

print(wallet_balances["minji"])           # 50000 — 키로 조회 / look up by key
wallet_balances["sora"] = 78000           # 새 사용자 추가 / add a new user
wallet_balances["minji"] += 5000          # 입금: 기존 값 수정 / deposit: update value
print(wallet_balances)                    # {'minji': 55000, 'junho': 32000, 'sora': 78000}

# (키, 값) 쌍으로 순회 / loop over (key, value) pairs
for name, balance in wallet_balances.items():
    print(f"{name}님의 잔액: {balance}원")
````

# 🎯 4. 집합 (Set) — 🎓 동아리 출석부

코딩 동아리의 출석 체크 기록에는 중복이 가득합니다 (체크인을 여러 번 한 사람들!). 중복 없는 모임, 그리고 "양일 모두 출석한 사람은?" 같은 질문 — 집합이 필요한 순간입니다.
The coding club's check-in log is full of duplicates (people who tapped in multiple times!). Unique membership, plus questions like "who attended BOTH days?" — that's set territory.

## 📋 핵심 연산 요약 / Core Operations

| 연산 / Operation | 문법 / Syntax | 설명 / Description |
|---|---|---|
| 생성 / Create | `s = {"민지", "준호"}` | 중괄호, **값만** (키:값 아님) |
| 빈 집합 / Empty set | `set()` | ⚠️ `{}`는 빈 **딕셔너리**! |
| 중복 제거 / Dedupe | `set(some_list)` | 리스트 → 집합 변환 |
| 추가 / Add | `s.add(x)` | 이미 있으면 변화 없음 |
| 삭제 / Remove | `s.remove(x)` | 없으면 `KeyError`! |
| 안전 삭제 / Safe remove | `s.discard(x)` | 없어도 에러 없음 |
| 합집합 / Union | `s1 \| s2` | 둘 중 하나라도 |
| 교집합 / Intersection | `s1 & s2` | 둘 다 |
| 차집합 / Difference | `s1 - s2` | s1에만 있는 것 |
| 포함 확인 / Membership | `x in s` | 매우 빠름! / very fast! |

> ⚠️ 집합은 **순서가 없습니다**. 인덱싱(`s[0]`)이 불가능하고, 출력 순서는 매번 다를 수 있어요!
> Sets are **unordered**: no indexing (`s[0]`), and print order may vary!

## 👀 따라해 보기 / Worked Examples

````python
# 예제: 중복 체크인 정리 / Example: dedupe the check-in log
checkin_log = ["민지", "준호", "민지", "소라", "준호"]   # 중복 체크인! / duplicate check-ins!
unique_members = set(checkin_log)
print(unique_members)         # {'민지', '준호', '소라'} — 순서는 다를 수 있음 / order may vary
print(len(unique_members))    # 3

# 집합 연산 / set operations
day1_attendance = {"민지", "준호", "소라"}
day2_attendance = {"준호", "태양"}
print(day1_attendance & day2_attendance)   # {'준호'} — 양일 모두 / both days
print(day1_attendance | day2_attendance)   # 합집합: 한 번이라도 출석 / attended at least once
````

# 🏆 보너스 미션 / Bonus Missions — 네 자료형 총출동!

이제 네 가지 자료형을 **함께** 사용해 봅시다. Wonder Pay의 거래 데이터가 들어왔습니다!
Time to use all four types **together**. Wonder Pay transaction data just arrived!

````python
# ─────────────────────────────────────────
# 🥉 [B1] 사용자별 거래 합계 / Per-user transaction totals
# 거래 기록(튜플 리스트)을 순회하며, 사용자별 합계 딕셔너리를 만드세요.
# Loop over the transactions (a list of tuples) and build a per-user totals dict.
# 힌트: 이름이 딕셔너리에 이미 있는지 in으로 확인! / Hint: check membership with `in`!
# ─────────────────────────────────────────
transactions = [("minji", 12000), ("junho", 5000), ("minji", 3000),
                ("sora", 20000), ("junho", 7000)]
user_totals = {}
for name, amount in transactions:
    if name in user_totals:
        user_totals[name] += amount
    else:
        user_totals[name] = amount
print(user_totals)

# 예상 출력 / Expected output:
# {'minji': 15000, 'junho': 12000, 'sora': 20000}
````

````python
# ─────────────────────────────────────────
# 🥈 [B2] 이틀 연속 거래 고객 / Customers who transacted both days
# 1) 각 날짜의 거래 기록에서 사용자 이름만 모은 집합을 만드세요.
# 2) 양일 모두 거래한 사용자와, 1일차에만 거래한 사용자를 구하세요.
# 1) Build a SET of user names from each day's transactions.
# 2) Find users who transacted BOTH days, and users who transacted ONLY on day 1.
# ─────────────────────────────────────────
day1_transactions = [("minji", 12000), ("junho", 5000), ("sora", 8000)]
day2_transactions = [("junho", 7000), ("sora", 20000), ("haneul", 4000)]

day1_users = set()
for name, amount in day1_transactions:
    day1_users.add(name)

day2_users = set()
for name, amount in day2_transactions:
    day2_users.add(name)

both_days = day1_users & day2_users
day1_only = day1_users - day2_users
print(f"양일 거래: {sorted(both_days)}")
print(f"1일차만 거래: {sorted(day1_only)}")

# 예상 출력 / Expected output:
# 양일 거래: ['junho', 'sora']
# 1일차만 거래: ['minji']
````

````python
# ─────────────────────────────────────────
# 🥇 [B3] VIP 리포트 / VIP Report
# 이틀 치 거래를 합쳐 사용자별 합계를 만들고,
# 총 거래액이 10000원 이상인 VIP를 이름순으로 출력하세요.
# Combine BOTH days, build per-user totals, then print VIPs
# (total >= 10000) in alphabetical order.
# 힌트: sorted(딕셔너리)는 키를 정렬한 리스트를 반환합니다!
# Hint: sorted(a_dict) returns a sorted list of its KEYS!
# ─────────────────────────────────────────
day1_transactions = [("minji", 12000), ("junho", 5000), ("sora", 8000)]
day2_transactions = [("junho", 7000), ("sora", 20000), ("haneul", 4000)]

all_transactions = day1_transactions + day2_transactions   # 리스트 연결 / concatenate lists
combined_totals = {}
for name, amount in all_transactions:
    if name in combined_totals:
        combined_totals[name] += amount
    else:
        combined_totals[name] = amount

print("=== 💎 Wonder Pay VIP 리포트 ===")
for name in sorted(combined_totals):          # 이름순 정렬 / alphabetical order
    if combined_totals[name] >= 10000:    # 경계 조건 주의! / watch the boundary!
        print(f"  💎 {name}: {combined_totals[name]}원")

# 예상 출력 / Expected output:
# === 💎 Wonder Pay VIP 리포트 ===
#   💎 junho: 12000원
#   💎 minji: 12000원
#   💎 sora: 28000원
````
