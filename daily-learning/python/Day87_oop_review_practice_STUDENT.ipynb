## 1. 캡슐화 / Encapsulation

**핵심 / Key points**
- `_single` (밑줄 하나): "내부용"이라는 **관례**일 뿐, 강제되지 않습니다.
  `_single` underscore is a **convention** meaning "internal use" — not enforced.
- `__double` (밑줄 둘): **이름 맹글링(name mangling)** 이 일어나 `_클래스명__속성` 으로 바뀝니다. 외부 접근이 더 어려워집니다.
  `__double` underscore triggers **name mangling** → renamed to `_ClassName__attr`, making outside access harder.

```python
# 관례적 비공개: 밑줄 하나 / Convention: single underscore
class Thermostat:
    def __init__(self, temp):
        self._temp = temp            # 내부용 속성 / internal attribute

    def get_temp(self):
        return self.___              # (1) 내부 속성 반환 / return internal attr

    def set_temp(self, value):
        if 10 <= value <= 30:
            self._temp = ___         # (2) 유효하면 갱신 / update if valid
        else:
            print("범위를 벗어났습니다 / Out of range")

t = Thermostat(22)
t.set_temp(25)
print(t.get_temp())                  # 예상 / Expected: 25
t.set_temp(50)                       # 예상 / Expected: 범위를 벗어났습니다 / Out of range
```

```python
# 이름 맹글링: 밑줄 둘 / Name mangling: double underscore
class Locker:
    def __init__(self, pin):
        self.__pin = pin             # 비공개 속성 / private attribute

    def check(self, guess):
        return guess == self.___     # (1) 비공개 속성과 비교 / compare to private

box = Locker(1234)
print(box.check(1234))               # 예상 / Expected: True
print(box.check(9999))               # 예상 / Expected: False
```

## 2. 상속 / Inheritance

**핵심 / Key points**
- 자식 클래스는 부모 클래스의 속성과 메서드를 **물려받습니다**.
  A child class **inherits** the parent's attributes and methods.
- `super()` 로 부모의 메서드(특히 `__init__`)를 호출합니다.
  Use `super()` to call the parent's methods (especially `__init__`).
- 자식이 부모 메서드를 **재정의(override)** 하거나, `super().메서드()` 로 **확장(extend)** 할 수 있습니다.
  A child can **override** a parent method, or **extend** it via `super().method()`.

```python
# 기본 상속 + super() / Basic inheritance + super()
class Animal:
    def __init__(self, name):
        self.name = name
    def speak(self):
        return "..."

class Dog(___):                      # (1) Animal 을 상속 / inherit from Animal
    def __init__(self, name, breed):
        ___.__init__(name)           # (2) 부모 생성자 호출 / call parent constructor
        self.breed = breed
    def speak(self):
        return "멍멍 / Woof"

d = Dog("Bba-dak", "Jindo")
print(d.name)                        # 예상 / Expected: Bba-dak
print(d.breed)                       # 예상 / Expected: Jindo
print(d.speak())                     # 예상 / Expected: 멍멍 / Woof
```

```python
# 부모 메서드 확장 / Extending a parent method
class Employee:
    def __init__(self, name, hours):
        self.name = name
        self.hours = hours
    def summary(self):
        return f"{self.name}: {self.hours}h"

class Manager(Employee):
    def __init__(self, name, hours, team_size):
        super().__init__(name, ___)  # (1) 부모에 name, hours 전달 / pass to parent
        self.team_size = team_size
    def summary(self):
        base = super().___()         # (2) 부모 메서드 재사용 / reuse parent method
        return f"{base} (team {self.team_size})"

m = Manager("Yuna", 40, 5)
print(m.summary())                   # 예상 / Expected: Yuna: 40h (team 5)
```

## 3. 다형성 / Polymorphism

**핵심 / Key points**
- **같은 이름의 메서드**가 객체의 타입에 따라 **다르게 동작**합니다.
  The **same method name** behaves **differently** depending on the object's type.
- 반복문에서 타입을 따지지 않고 동일한 메서드를 호출할 수 있습니다.
  In a loop, you can call the same method without checking the type.
- **덕 타이핑(duck typing)**: "그 메서드가 있으면 그냥 호출한다." 타입보다 동작이 중요합니다.
  **Duck typing**: "if it has the method, just call it." Behavior matters more than type.

```python
# 메서드 재정의 + 다형성 반복 / Overriding + polymorphic loop
class Shape:
    def area(self):
        return 0

class Rectangle(Shape):
    def __init__(self, w, h):
        self.w = w
        self.h = h
    def area(self):
        return self.w * ___          # (1) 넓이 계산 / compute area

class Circle(Shape):
    def __init__(self, r):
        self.r = r
    def area(self):
        return 3.14 * self.r * self.r

shapes = [Rectangle(3, 4), Circle(2)]
for s in shapes:
    print(s.___())                   # (2) 각 도형의 area 호출 / call each area
# 예상 / Expected: 12  그리고/then 12.56
```

```python
# 덕 타이핑 / Duck typing
class Dog:
    def make_sound(self):
        return "멍멍 / Woof"

class Bell:
    def make_sound(self):
        return "딸랑 / Ring"

def play(item):
    return item.___()                # (1) 타입 상관없이 호출 / call regardless of type

print(play(Dog()))                   # 예상 / Expected: 멍멍 / Woof
print(play(Bell()))                  # 예상 / Expected: 딸랑 / Ring
```

## 4. 던더 메서드 / Dunder methods

**핵심 / Key points**
- **던더(dunder)** = **d**ouble **under**score, 즉 `__name__` 형태의 특수 메서드입니다. 파이썬이 자동으로 호출합니다.
  A **dunder** is a `__name__` special method that Python calls automatically.
- `__str__`: `print()` 용 사람 친화적 문자열 / human-friendly string for `print()`
- `__repr__`: 개발자/디버깅용 문자열 (리스트 안에서 표시될 때 사용) / developer string (used inside lists)
- `__eq__` (`==`), `__lt__` (`<`), `__len__` (`len()`), `__getitem__` (`obj[i]`), `__add__` (`+`)

```python
# __str__ : print()가 사용하는 메서드 / used by print()
class Book:
    def __init__(self, title, pages):
        self.title = title
        self.pages = pages
    def ___(self):                   # (1) print()가 호출하는 던더 / dunder print() uses
        return f"'{self.title}' ({self.pages}p)"

b = Book("파이썬 입문", 320)
print(b)                             # 예상 / Expected: '파이썬 입문' (320p)
print(str(b))                        # 예상 / Expected: '파이썬 입문' (320p)
```

```python
# __eq__ : == 비교를 직접 정의 / define == comparison
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y
    def __eq__(self, other):
        return self.x == other.x and self.y == ___   # (1) y 좌표도 비교 / compare y too

p1 = Point(1, 2)
p2 = Point(1, 2)
p3 = Point(3, 4)
print(p1 == p2)                      # 예상 / Expected: True
print(p1 == p3)                      # 예상 / Expected: False
```

```python
# __len__ 과 __getitem__ : len() 과 인덱싱 지원 / support len() and indexing
class Playlist:
    def __init__(self, songs):
        self.songs = songs
    def __len__(self):
        return len(self.___)         # (1) 곡 개수 / number of songs
    def __getitem__(self, index):
        return self.songs[___]       # (2) 인덱스로 곡 접근 / access by index

pl = Playlist(["A", "B", "C"])
print(len(pl))                       # 예상 / Expected: 3
print(pl[0])                         # 예상 / Expected: A
print(pl[2])                         # 예상 / Expected: C
```

```python
# __add__ : + 연산자 정의 / define the + operator
class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y
    def __add__(self, other):
        return Vector(self.x + other.x, self.y + ___)  # (1) y 성분 합 / add y parts
    def __str__(self):
        return f"({self.x}, {self.y})"

v = Vector(1, 2) + Vector(3, 4)
print(v)                             # 예상 / Expected: (4, 6)
```

## 5. 추상 베이스 클래스 / Abstract base classes (`abc`)

**핵심 / Key points**
- `from abc import ABC, abstractmethod` 로 가져옵니다.
  Import with `from abc import ABC, abstractmethod`.
- `ABC` 를 상속하고 `@abstractmethod` 를 붙이면, 그 메서드는 **자식이 반드시 구현**해야 합니다.
  Inheriting `ABC` with `@abstractmethod` forces **subclasses to implement** that method.
- 추상 메서드를 구현하지 않으면 그 클래스는 **인스턴스화할 수 없습니다** (`TypeError`).
  An abstract class **cannot be instantiated** until all abstract methods are implemented (`TypeError`).

```python
# 추상 베이스 클래스 정의 / Define an abstract base class
from abc import ABC, abstractmethod

class Sender(___):                   # (1) 추상 베이스 클래스 상속 / inherit abstract base
    @___                             # (2) 추상 메서드 표시 / mark as abstract
    def send(self, message):
        pass

class EmailSender(Sender):
    def send(self, message):
        return f"📧 {message}"

e = EmailSender()
print(e.send("안녕 / Hi"))           # 예상 / Expected: 📧 안녕 / Hi
```

**출력을 예측한 뒤 실행하세요 / Predict the output, then run.**
*힌트 / Hint: 추상 메서드를 구현하지 않은 클래스를 인스턴스화하면 어떻게 될까요? / What happens if you instantiate a class with an unimplemented abstract method?*

```python
from abc import ABC, abstractmethod

class Animal(ABC):
    @abstractmethod
    def sound(self):
        pass

try:
    a = Animal()
except TypeError:
    print("추상 클래스는 인스턴스화 불가 / Cannot instantiate abstract class")

class Dog(Animal):
    def sound(self):
        return "멍멍"

print(Dog().sound())
```

## 6. 통합 챌린지 / Integration challenge

지금까지 배운 것을 한 번에 / Everything together: **ABC + 추상 메서드 + 상속 + 다형성 + `__str__`**

도서관 자료 시스템을 만듭니다. 모든 자료는 `loan_period`(대출 기간)를 가져야 하며,
`__str__` 은 그 기간을 사용해 표시합니다.
A library-item system: every item must define `loan_period`, and `__str__` uses it to display.

```python
from abc import ABC, abstractmethod

class LibraryItem(ABC):
    def __init__(self, title):
        self.title = title

    @abstractmethod
    def loan_period(self):           # 자식이 반드시 구현 / subclasses must implement
        pass

    def __str__(self):
        return f"{self.title} ({self.loan_period()}일 / days)"

class PaperBook(LibraryItem):
    def loan_period(self):
        return ___                   # (1) 종이책은 14일 / paper book: 14 days

class Dvd(LibraryItem):
    def loan_period(self):
        return 7

items = [PaperBook("파이썬"), Dvd("영화")]
for it in items:
    print(it)
# 예상 / Expected:
# 파이썬 (14일 / days)
# 영화 (7일 / days)
```
