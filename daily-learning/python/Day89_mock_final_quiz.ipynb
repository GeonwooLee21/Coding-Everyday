**B1. 버그 찾기 — 이름 서식 / Find the bug — formatting names** &nbsp;&nbsp;`(5점 / pts)`

🎯 의도: 각 이름을 Title Case로 바꿔 반환
*Intent: return each name in Title Case*

🐞 버그 / Bug: ________________________________________
🔧 수정 방법 / Fix: ____________________________________

```python
def format_names(raw):
    names = raw.split(",")
    for name in names:
        name = name.title()
    return names

print(format_names("kim minji,lee junho"))
# 기대 / Expected: ['Kim Minji', 'Lee Junho']
```

**B2. 버그 찾기 — 장르 세기 / Find the bug — counting genres** &nbsp;&nbsp;`(5점 / pts)`

🎯 의도: 각 장르가 몇 번 대출됐는지 세기
*Intent: count how many times each genre was borrowed*

🐞 버그 / Bug: ________________________________________
🔧 수정 방법 / Fix: ____________________________________

```python
def count_genres(genres):
    counts = {}
    for g in genres:
        counts[g] += 1
    return counts

print(count_genres(["fiction", "fiction", "poetry"]))
# 기대 / Expected: {'fiction': 2, 'poetry': 1}
```

**B3. 버그 찾기 — 고유 방문자 / Find the bug — unique visitors** &nbsp;&nbsp;`(5점 / pts)`

🎯 의도: 고유한 [카드, 게이트] 쌍의 개수 반환
*Intent: return the number of unique [card, gate] pairs*

🐞 버그 / Bug: ________________________________________
🔧 수정 방법 / Fix: ____________________________________

```python
def unique_visitors(log):
    seen = set()
    for entry in log:
        seen.add(entry)
    return len(seen)

print(unique_visitors([[1, 2], [1, 2], [3]]))
# 기대 / Expected: 2
```

**B4. 버그 찾기 — 식비 합계 / Find the bug — total spent** &nbsp;&nbsp;`(5점 / pts)`

🎯 의도: 모든 가격을 더하기
*Intent: add up every price*

🐞 버그 / Bug: ________________________________________
🔧 수정 방법 / Fix: ____________________________________

```python
def total_spent(prices):
    total = 0
    for i in range(len(prices) - 1):
        total += prices[i]
    return total

print(total_spent([3000, 2000, 1500]))
# 기대 / Expected: 6500
```

**B5. 버그 찾기 — 짐 싸기 목록 / Find the bug — packing list** &nbsp;&nbsp;`(5점 / pts)`

🎯 의도: 호출할 때마다 새 가방에 물건 하나만 담기
*Intent: each call starts a fresh bag with just that item*

🐞 버그 / Bug: ________________________________________
🔧 수정 방법 / Fix: ____________________________________

```python
def add_item(item, bag=[]):
    bag.append(item)
    return bag

print(add_item("pillow"))
print(add_item("blanket"))
# 기대 / Expected: ['pillow'] 그리고 / then ['blanket']
```

**B6. 버그 찾기 — 연체 도서 확인 / Find the bug — overdue check** &nbsp;&nbsp;`(5점 / pts)`

🎯 의도: 14일 넘게 연체된 책이 하나라도 있으면 True
*Intent: True if ANY book is more than 14 days overdue*

🐞 버그 / Bug: ________________________________________
🔧 수정 방법 / Fix: ____________________________________

```python
def has_overdue(days_list):
    for days in days_list:
        if days > 14:
            return True
        else:
            continue
    return False

print(has_overdue([10, 20, 5]))
# 기대 / Expected: True
```

**B7. 버그 찾기 — 식권 카드 클래스 / Find the bug — meal-card class** &nbsp;&nbsp;`(5점 / pts)`

🎯 의도: 잔액으로 카드를 만들고 결제 후 잔액 반환
*Intent: build a card with a balance, pay, and return the remaining balance*

🐞 버그 / Bug: ________________________________________
🔧 수정 방법 / Fix: ____________________________________

```python
class MealCard:
    def __init__(self, balance):
        balance = balance

    def pay(self, amount):
        self.balance -= amount
        return self.balance

card = MealCard(10000)
print(card.pay(3000))
# 기대 / Expected: 7000
```

**B8. 버그 찾기 — 요금 분할 / Find the bug — splitting a fare** &nbsp;&nbsp;`(5점 / pts)`

🎯 의도: 인원이 0명이면 0을 반환하고, 그 외에는 1인당 요금 반환
*Intent: return 0 when there are 0 riders, otherwise the per-rider fare*

🐞 버그 / Bug: ________________________________________
🔧 수정 방법 / Fix: ____________________________________

```python
def divide_fare(total, riders):
    try:
        result = total / riders
    except ValueError:
        result = 0
    return result

print(divide_fare(9000, 0))
# 기대 / Expected: 0
```
